# Subplots (Multiple Charts in One Figure)

In professional dashboards, you rarely show just one chart. You usually want to see a line chart, a bar chart, and a histogram next to each other so you can compare different perspectives of your data.

In Matplotlib, we do this using the `plt.subplots()` function. It helps us split our plotting window into rows and columns of grids.

Let's define two extremely important terms:
*   **Figure (fig)**: The entire outer window or canvas. It holds all your charts.
*   **Axes (ax / plural: axes)**: An individual chart or plot inside the Figure.
    *   *Note*: Do not confuse "axes" with the X and Y coordinate lines. In Matplotlib code, `ax` represents the actual plot object.


### 2. Simple Explanation with an Everyday Analogy
Imagine a **wooden picture frame** hanging on your wall:
*   The entire **wooden frame** is the **Figure**.
*   Inside the frame, there are **4 small photo slots** (a 2x2 grid). Each slot holds a separate photo. Each photo slot is an **Axis** (or subplot).
*   You can choose to paint a red line in slot `[0, 0]` (top-left) and stick a bar chart in slot `[1, 1]` (bottom-right).


### Code Examples

When you call `plt.subplots(rows, cols)`, it returns a **tuple** containing two things:
1. The **Figure** object.
2. An array of **Axes** objects (which can be 1D or 2D depending on your grid shape).

#### Creating a 2x2 Grid of Subplots
Let's build a beautiful 2x2 grid. We will unpack the axes array using row and column indices:

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Prepare some mathematical data
x = np.arange(1, 6) # [1, 2, 3, 4, 5]

# 2. Create the subplots grid (2 rows, 2 columns)
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))

# 3. Plot on Top-Left [0, 0]
axes[0, 0].plot(x, x * 2, color="red", marker="o")
axes[0, 0].set_title("Linear (y = x * 2)") # Note: We use .set_title() instead of .title() [35]

# 4. Plot on Top-Right [0, 1]
axes[0, 1].plot(x, x ** 2, color="blue", marker="s")
axes[0, 1].set_title("Square (y = x^2)") # [35]

# 5. Plot on Bottom-Left [1, 0]
axes[1, 0].plot(x, x ** 3, color="green", marker="^")
axes[1, 0].set_title("Cubic (y = x^3)") # [35, 36]

# 6. Plot on Bottom-Right [1, 1]
# Let's make this one a bar chart! [36]
axes[1, 1].bar(x, x * 10, color="purple")
axes[1, 1].set_title("Bar Chart (y = x * 10)") # [36]

# 7. Use tight_layout() to prevent overlapping labels and titles! [36]
plt.tight_layout() # [36]

# 8. Save the entire multi-plot figure [34]
# plt.savefig("/workspace/scratch/subplots_2x2.png", dpi=150)
plt.show()
plt.close()
print("2x2 subplots grid created!")
```

### 4. Common Mistakes Beginners Make
1.  **Using `plt.title()` inside subplots**:
    *   If you write `plt.title()`, Matplotlib will only add the title to the *very last* active plot.
    *   When working with individual subplots, you **must use the object-oriented method**: `axes[row, col].set_title("My Title")`. The same goes for labels: use `.set_xlabel()` and `.set_ylabel()`.
2.  **Forgetting `plt.tight_layout()`**: Without this, the title of the bottom plots will overlap with the X-axis of the top plots, making your charts look messy and unreadable.
3.  **Mismatched Indices in 1D vs 2D Grids**:
    *   If you create a 1D grid like `plt.subplots(1, 2)` (1 row, 2 columns), `axes` is a **1D array**. You access them using `axes[0]` and `axes[1]`.
    *   If you create a 2D grid like `plt.subplots(2, 2)`, `axes` is a **2D array**. You must use two indices: `axes[row, col]`.


#### Exercise 1 (Medium)
Create a 1D row of 2 subplots (1 row, 2 columns):
*   Left Subplot (`axes[0]`): Plot a red line showing `y = [5, 10, 15]`.
*   Right Subplot (`axes[1]`): Plot a blue dashed line showing `y = [15, 10, 5]`.
Use `plt.tight_layout()` and save the output.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3])

fig, axes = plt.subplots(1, 2, figsize=(8, 4))

# Left plot
axes[0].plot(x, [5, 10, 15], color="red", marker="o")
axes[0].set_title("Going Up")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")

# Right plot
axes[1].plot(x, [15, 10, 5], color="blue", linestyle="--", marker="s")
axes[1].set_title("Going Down")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Y")

plt.tight_layout()
# plt.savefig("/workspace/scratch/ex_subplots_1d.png")
plt.show()
plt.close()
```
