# Titles, Labels, Ticks, and Gridlines

A chart without labels is just a random line. Nobody will know what it means!

In this module, you will learn how to:
*   Add a **Title** to your chart and customize its font size and weight.
*   Add **Labels** to the horizontal (X) and vertical (Y) axes.
*   Customize **Ticks** (the coordinate markings on the axes).
*   Add **Gridlines** to make the values much easier to read.

### Simple Explanation with an Everyday Analogy
Imagine you are looking at a **thermometer**:
*   The title at the top says: "Room Temperature".
*   The side label says: "Degrees Celsius".
*   The small lines at $10^\circ$, $20^\circ$, $30^\circ$ are the **ticks**.
*   The light lines stretching across the background are the **gridlines**, helping you see exactly where the mercury column ends.

These elements turn a simple drawing into a precise scientific measurement tool!

#### Technical Term Explainers:
*   **Ticks**: The scale markings (numbers or labels) on the X and Y axes.
*   **Gridlines**: The background reference lines that help align data points to axis numbers.


#### Complete Labeled Plot with Gridlines
Let's build a complete, highly professional chart.

```python
import numpy as np
import matplotlib.pyplot as plt

months = np.array(['Jan', 'Feb', 'Mar', 'Apr'])
bugs_fixed = np.array([12, 28, 19, 34])

plt.plot(months, bugs_fixed, marker='s', color='#FF5733', linewidth=3) # 's' is square marker

# 1. Add a Title with custom font, size, weight, and color
plt.title("Bugs Fixed in Q1", fontsize=20, fontname='Arial', fontweight='bold', color='#2C3E50')

# 2. Add X and Y Axis Labels
plt.xlabel("Months", fontsize=12, fontweight='bold', color='#34495E')
plt.ylabel("Number of Bugs", fontsize=12, fontweight='bold', color='#34495E')

# 3. Add background Gridlines
# - axis='y' means only draw horizontal lines
# - linestyle='dashed' makes lines dashed
# - color='lightgray' keeps them subtle
plt.grid(axis='y', linestyle='dashed', color='lightgray', linewidth=1)

plt.show()
```

#### Customizing Axis Ticks
Sometimes, Matplotlib shows decimal numbers on the axis (like `1.5` or `2.5`) when you only want integer years. You can force Matplotlib to use specific ticks!

```python
import numpy as np
import matplotlib.pyplot as plt

years = np.array([2023, 2024, 2025, 2026])
sales = np.array([100, 150, 130, 200])

plt.plot(years, sales, marker='o')

# Force X-axis ticks to display ONLY our exact years
plt.xticks(years)

# Customize tick parameters (make both ticks colored red for visual demo)
plt.tick_params(axis='both', colors='red', labelsize=12)

plt.show()
```

### 4. Common Mistakes Beginners Make
1.  **Over-cluttering the grid**: Adding a dark, thick black grid (`plt.grid(color='black', linewidth=3)`). This makes the grid stand out more than the actual data! Keep gridlines light and subtle (like light gray).
2.  **Using `plt.title()` after `plt.show()`**: Writing `plt.show()` first and then configuring titles/labels. Python executes code top-to-bottom. Once `plt.show()` runs, the chart window is closed and a new blank canvas is started. Always put customization code **before** `plt.show()`!
3.  **Forgetting that `axis` in `plt.grid()` takes string values**: Writing `plt.grid(axis=0)` instead of `plt.grid(axis='x')` or `plt.grid(axis='both')`.

Create a simple plot of any dataset. Add the title "My First Chart" with a font size of 18, and set both axis labels.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3])
y = np.array([10, 20, 30])

plt.plot(x, y)
plt.title("My First Chart", fontsize=18)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

#### Exercise 2
Create a plot of 5 data points.
1. Add custom light-gray dashed gridlines only along the X-axis (vertical lines).
2. Force the X-axis ticks to show only the numbers `[1, 2, 3, 4, 5]`.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([15, 22, 18, 30, 25])

plt.plot(x, y)

# Set custom X ticks
plt.xticks(x)

# Add vertical gridlines (along X axis)
plt.grid(axis='x', color='lightgray', linestyle='--')

plt.show()
```
