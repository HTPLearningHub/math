# Cleaning Text and Standardizing Strings

Since real-world datasets often involve text entered by humans, string columns are usually filled with inconsistencies. These include:
* **Inconsistent Capitalization**: Having `'python'`, `'Python'`, and `'PYTHON'` in the same column. Pandas will treat these as three completely different categories!
* **Accidental Spaces**: Extra spaces at the beginning or end of strings (e.g., `' Latte '` instead of `'Latte'`). These spaces are invisible but prevent direct matching and grouping.
* **Unwanted Characters**: Strange characters (like `/`, `.`, `_`, or `$`) embedded in the text.

Pandas allows you to clean entire columns of text easily by calling the **`.str`** attribute before applying a string method.


### 2. Plain English Explanation & Real-Life Analogy
Imagine you are sorting a stack of physical paper files.
* Some names are written in ALL CAPS, some are lowercase, and some have messy leading spaces.
* To organize them, you take three actions:
  1. **Standardize Case**: You rewrite every folder name in proper title case (first letter capitalized).
  2. **Trimming**: You cut off any useless blank margins (leading or trailing spaces).
  3. **Replacing**: You erase any accidental typos or ink splatters (unwanted characters) with white correction tape.

###  Code Examples

Let's create a messy DataFrame containing names of programming languages with capitalization issues, accidental spaces, and unwanted characters.



````python
import pandas as pd

data = {
    'Language': ['  typescript ', 'PYTHON  ', 'java_script', '  typescript ', 'python'],
    'Developer_ID': [101, 102, 103, 104, 105]
}

df = pd.DataFrame(data)
print(df)
````

#### Capitalization (`.str.lower()`, `.str.upper()`, `.str.title()`)
To fix case inconsistencies, use lowercase, uppercase, or title case.

````python
# Convert to lower case
df['Language_Lower'] = df['Language'].str.lower()

# Convert to upper case
df['Language_Upper'] = df['Language'].str.upper()

print(df[['Language', 'Language_Lower', 'Language_Upper']])
````


#### Trimming Spaces (`.str.strip()`, `.str.lstrip()`, `.str.rstrip()`)
* **`.str.strip()`**: Removes spaces from both the left and right sides.
* **`.str.lstrip()`**: Removes spaces only from the left side.
* **`.str.rstrip()`**: Removes spaces only from the right side.


````python
# Strip spaces from both ends
df['Language_Clean'] = df['Language'].str.strip()
print(df[['Language', 'Language_Clean']])
````

#### C) Replacing Characters (`.str.replace()`)
You can replace characters or substrings inside a column using `.str.replace()`. For example, let's replace the underscore `_` in `java_script` with a space.

````python
# Replace underscores with spaces
df['Language_Clean'] = df['Language_Clean'].str.replace('_', '')
print(df[['Language_Clean']])
````

### 4. Common Pitfalls
1. **Calling String Methods on Numeric Columns**: Trying to run `.str` methods on integer or float columns (e.g., `df['Developer_ID'].str.lower()`) will result in an `AttributeError`. String methods can only be applied to string (object) columns.
2. **Forgetting to Chain `.str`**: You cannot write `df['Language'].lower()`. You must explicitly call `.str` first: `df['Language'].str.lower()`.

#### Exercise 1 (Easy)
You have a DataFrame of customer emails:
```python
emails = pd.DataFrame({
    'Email': ['  John@Gmail.com ', 'Spongebob@yahoo.com  ', '  PATRICK@GMAIL.COM']
})
```
Write Pandas code to:
1. Strip all leading and trailing white spaces.
2. Convert all emails to lowercase to make them fully standardized.


````python
import pandas as pd

emails = pd.DataFrame({
    'Email': ['  John@Gmail.com ', 'Spongebob@yahoo.com  ', '  PATRICK@GMAIL.COM']
})

# Strip spaces and convert to lowercase in a chained operation
emails['Email_Clean'] = emails['Email'].str.strip().str.lower()
print(emails)
````

#### Exercise 2 (Medium)
You have a DataFrame of salaries containing dollar signs and commas as strings:
```python
salaries = pd.DataFrame({
    'Employee': ['Alex', 'John', 'Sarah'],
    'Salary_Str': ['$85,000', '$110,000', '$75,000']
})
```
1. Remove the dollar sign `$` and the comma `,` from the `Salary_Str` column.
2. Convert the cleaned column to integer type (`int`) so you can perform calculations.

````python
import pandas as pd

salaries = pd.DataFrame({
    'Employee': ['Alex', 'John', 'Sarah'],
    'Salary_Str': ['$85,000', '$110,000', '$75,000']
})

# 1. Clean the string column by replacing symbols
salaries['Salary_Int'] = salaries['Salary_Str'].str.replace('$', '').str.replace(',', '')

# 2. Convert string to integer type
salaries['Salary_Int'] = salaries['Salary_Int'].astype(int)

# 3. Test calculation (calculate a 10% raise)
salaries['Salary_New'] = salaries['Salary_Int'] * 1.10

print(salaries)
````
