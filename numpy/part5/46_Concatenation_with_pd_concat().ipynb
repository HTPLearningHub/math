# Concatenation with `pd.concat()`

Concatenation is the process of gluing or stacking DataFrames together along an axis (either row-wise or column-wise).
The most common use case is **vertical stacking** (axis=0). For example, if you receive monthly sales reports (January data, February data, March data) with the exact same columns, you want to stack them vertically to create a single, master DataFrame for the whole year.

You can also concatenate **horizontally** (axis=1). This is useful when you have completely different columns of data for the exact same rows (matching indices), and you want to place them side-by-side.

### 2. Clear Explanation & Real-World Analogy
*   **Vertical Concatenation (axis=0)**: Imagine you are a teacher collecting exam papers. You have a stack of papers from Class A (DataFrame 1) and a stack of papers from Class B (DataFrame 2). To grade them together, you simply put Class B's stack directly underneath Class A's stack. They have the same columns (Name, Score), so they fit perfectly.

*   **Horizontal Concatenation (axis=1)**: Imagine you have a sheet containing names and ages of 4 students (DataFrame 1). You also have another sheet containing the phone numbers of the *exact same 4 students* in the exact same order (DataFrame 2). You lay these two sheets side-by-side on your desk to read them as one wide sheet.

### Code Examples

First, let's create two simple DataFrames containing employee data for January and February:



````python
import pandas as pd

# DataFrame 1: January Hires
jan_df = pd.DataFrame({
    'Name': ['Alice', 'Bob'],
    'Role': ['TypeScript Developer', 'Python Developer'],
    'Salary': [85000, 95000]
})

# DataFrame 2: February Hires
feb_df = pd.DataFrame({
    'Name': ['Charlie', 'David'],
    'Role': ['Angular Specialist', 'Data Scientist'],
    'Salary': [90000, 110000]
})

print("--- January Hires ---")
print(jan_df)
print("--- February Hires ---")
print(feb_df)
````

#### A) Stacking Vertically (`axis=0`)
By default, `pd.concat()` stacks DataFrames vertically. We pass our DataFrames as a list inside the function:

````python
# Vertical Concat (Default)
combined_vertical = pd.concat([jan_df, feb_df])
print(combined_vertical)
````

> ⚠️ **Notice the Index Issue:** The row indices repeated (`0, 1, 0, 1`) because Pandas preserved the original indices of both DataFrames. To fix this and generate a clean, continuous index (`0, 1, 2, 3`), pass `ignore_index=True`:

````python
# Vertical Concat with clean index
combined_clean = pd.concat([jan_df, feb_df], ignore_index=True)
print(combined_clean)
````

#### Placing Side-by-Side Horizontally (`axis=1`)
Now, let's say you have some extra info (such as phone numbers) for January hires, and the row indices match exactly:


````python
phone_df = pd.DataFrame({
    'Phone': ['555-0192', '555-0143']
})

# Horizontal Concat (axis=1)
combined_horizontal = pd.concat([jan_df, phone_df], axis=1)
print(combined_horizontal)
````


### Common Pitfalls to Avoid
1.  **Mismatching Column Names**: If you concatenate vertically, but column names do not match exactly (e.g., `'Salary'` vs. `'salary_yearly'`), Pandas will not align them. It will create separate columns for both and fill the missing cells with `NaN`. Always standardize column headers before concatenating.
2.  **Forgetting to Pass as a List**: Writing `pd.concat(jan_df, feb_df)` will result in a `TypeError`. You must always enclose the DataFrames in a Python list: `pd.concat([jan_df, feb_df])`.
