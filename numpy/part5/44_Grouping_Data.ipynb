# Grouping Data

Basic aggregations are helpful, but they treat the entire column as one group. In real-world analysis, you usually want to compare subgroups. For example: "What is the average salary **grouped by** programming language?"

To do this, Pandas uses a concept called **Split-Apply-Combine**:
1.  **Split**: Divide the DataFrame into groups based on a column's unique values.
2.  **Apply**: Apply a statistical function (like sum or mean) to each group independently.
3.  **Combine**: Merge the results back into a single, clean summary DataFrame.

### Analogy: The Laundry Sorting
Imagine you have a huge pile of unsorted laundry.
1.  **Split**: You sort the clothes into three baskets: Whites, Darks, and Colors.
2.  **Apply**: You weigh each basket separately (Whites = 5kg, Darks = 8kg, Colors = 4kg).
3.  **Combine**: You write down a single summary sheet:
    *   Whites: 5kg
    *   Darks: 8kg
    *   Colors: 4kg

This is exactly how `.groupby()` works under the hood.

### Code Examples

Let's use our developer DataFrame to group by the `Language` column and calculate group statistics.

#### Standard Grouping with an Aggregation
To group your data, call `.groupby()`, specify the column you want to group by, select the target numeric column you want to analyze, and append your aggregation function:


````python
import pandas as pd

data = {
    'Name': ['Alex', 'John', 'Sarah', 'Emily', 'David'],
    'Language': ['TypeScript', 'Python', 'JavaScript', 'Python', 'TypeScript'],
    'Experience': [3, 5, 2, 8, 1],
    'Salary': [85000, 110000, 75000, 145000, 60000]
}

df = pd.DataFrame(data)

# Group by Language and calculate the average (mean) salary for each language
language_salary = df.groupby('Language')['Salary'].mean()
print(language_salary)
````

#### Understanding the GroupBy Object
If you run `df.groupby('Language')` without adding an aggregation function, Pandas will return a generic object:

````python
grouped_obj = df.groupby('Language')
print(grouped_obj)
````

*Note:* A GroupBy object is lazy. It does not perform any calculations until you tell it **what** aggregation function to apply (like `.mean()`, `.sum()`, or `.count()`).

#### Grouping by Multiple Columns
You can group by more than one category by passing a list of column names to `.groupby()`:


````python
# Let's add a 'Role' column to our data for this example
df['Role'] = ['Junior', 'Senior', 'Junior', 'Senior', 'Junior']

# Group by both Language AND Role, then calculate median salaries
multi_group = df.groupby(['Language', 'Role'])['Salary'].median()
print(multi_group)
````

### Common Pitfalls for Beginners
1.  **Forgetting the Aggregation**: Writing `df.groupby('Language')['Salary']` and expecting a DataFrame. Remember, you must chain an aggregation function (like `.mean()`) to get a visible result.
2.  **Incorrect Syntax Order**: Always keep this order in mind:
    `df.groupby('COLUMN_TO_GROUP')['NUMERIC_COLUMN_TO_ANALYZE'].AGGREGATION_FUNCTION()`


#### Exercise 1 (Intermediate)
You have a dataset of fruit sales:
```python
sales_data = pd.DataFrame({
    'Fruit': ['Apple', 'Banana', 'Apple', 'Banana', 'Apple'],
    'Quantity': [10, 15, 7, 20, 8],
    'Price_Per_Unit': [1.2, 0.5, 1.2, 0.5, 1.2]
})
```
Write Pandas code to:
1. Group by `Fruit` and find the total `Quantity` sold for each fruit.
2. Group by `Fruit` and find the average `Quantity` sold per transaction.



````python
import pandas as pd
sales_data = pd.DataFrame({
    'Fruit': ['Apple', 'Banana', 'Apple', 'Banana', 'Apple'],
    'Quantity': [10, 15, 7, 20, 8],
    'Price_Per_Unit': [1.2, 0.5, 1.2, 0.5, 1.2]
})

# 1. Total Quantity Sold
total_qty = sales_data.groupby('Fruit')['Quantity'].sum()
print("--- Total Quantity Sold ---")
print(total_qty)

# 2. Average Quantity Per Sale
avg_qty = sales_data.groupby('Fruit')['Quantity'].mean()
print("--- Average Quantity Per Sale ---")
print(avg_qty)
````
