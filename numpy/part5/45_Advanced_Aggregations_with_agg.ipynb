 # Advanced Aggregations with `.agg()`

Sometimes, calculating just the mean or the sum is not enough. You might want to calculate **both** the minimum and maximum salary at the same time, or apply different statistical operations to different columns (e.g., sum up total quantity, but find the average price).

To achieve this level of flexibility, Pandas provides the highly versatile **`.agg()`** (Aggregation) method. This method allows you to:
*   Apply multiple statistical functions to a single column simultaneously.
*   Use a dictionary to map specific aggregation rules to different columns.
*   Rename resulting columns dynamically during aggregation.

### Analogy: The Custom Request Menu
Imagine ordering food at a custom restaurant.
*   A basic aggregation is like everyone at the table ordering the same dish (e.g., "everyone gets soup").
*   Using `.agg()` with a list is like ordering multiple starters for yourself ("I want the soup, the salad, AND the main course").
*   Using `.agg()` with a dictionary is like placing custom orders for each guest ("The kids get burgers, the adults get steak, and we want average-sized portions of fries").

### Code Examples

We will continue using our developer dataset `df` to demonstrate advanced aggregations.

#### Applying Multiple Aggregations to One Column
You can pass a list of string functions (such as `'min'`, `'max'`, `'mean'`, `'count'`) into `.agg()` to compute multiple metrics at once:



````python
import pandas as pd

data = {
    'Name': ['Alex', 'John', 'Sarah', 'Emily', 'David'],
    'Language': ['TypeScript', 'Python', 'JavaScript', 'Python', 'TypeScript'],
    'Experience': [3, 5, 2, 8, 1],
    'Salary': [85000, 110000, 75000, 145000, 60000]
}

df = pd.DataFrame(data)

# Group by Language, then find min, max, and mean of salaries
advanced_salary = df.groupby('Language')['Salary'].agg(['min', 'max', 'mean'])
print(advanced_salary)
````

#### Applying Different Aggregations to Different Columns

By passing a dictionary into `.agg()`, you can define custom metrics for each column. The key is the column name, and the value is the aggregation function (or list of functions):

````python
# Group by Language: find total Experience sum, but average (mean) Salary
custom_rules = df.groupby('Language').agg({
    'Experience': 'sum',
    'Salary': 'mean'
})
print(custom_rules)
````

#### Named Aggregations (Custom Column Names)
To prevent your aggregated DataFrame from having generic column headers (like `mean` or `sum`), you can use **Named Aggregations**. This allows you to name the output columns directly:

````python
# Group by Language and define clean, custom column headers
named_agg = df.groupby('Language').agg(
    Total_Devs=('Name', 'count'),
    Average_Salary=('Salary', 'mean'),
    Max_Experience=('Experience', 'max')
)
print(named_agg)
````

### Common Pitfalls for Beginners
1.  **Passing incorrect string functions**: Typing `'average'` instead of `'mean'` inside `.agg()`. Pandas only recognizes specific built-in keywords as strings (e.g., `'mean'`, `'sum'`, `'min'`, `'max'`, `'std'`, `'count'`).
2.  **Multi-index columns confusion**: When you apply multiple functions to columns, Pandas creates a hierarchical column index (MultiIndex). If you want to flatten these column headers back to a single level, you can rename them or use Named Aggregations (shown in example C above).

#### Exercise 1 (Hard)
Using the previous `sales_data` DataFrame:
```python
sales_data = pd.DataFrame({
    'Fruit': ['Apple', 'Banana', 'Apple', 'Banana', 'Apple'],
    'Quantity': [10, 15, 7, 20, 8],
    'Price_Per_Unit': [1.2, 0.5, 1.2, 0.5, 1.2]
})
```
Write Pandas code to group by `Fruit` and:
1. Find the total sum of `Quantity` AND the average (mean) of `Price_Per_Unit` using a dictionary.
2. Flatten the columns or use Named Aggregations so the final DataFrame has two columns called `Total_Sold` and `Avg_Price_Per_Unit`.


````python
sales_data = pd.DataFrame({
    'Fruit': ['Apple', 'Banana', 'Apple', 'Banana', 'Apple'],
    'Quantity': [10, 15, 7, 20, 8],
    'Price_Per_Unit': [1.2, 0.5, 1.2, 0.5, 1.2]
})

final_report = sales_data.groupby('Fruit').agg(
    Total_Sold=('Quantity', 'sum'),
    Avg_Price_Per_Unit=('Price_Per_Unit', 'mean')
)
print(final_report)
````

*Explanation:* Named aggregation is the cleanest way to perform multi-column calculations. We specified exactly which column to target (`Quantity` and `Price_Per_Unit`), which function to use (`sum` and `mean`), and assigned them elegant, highly descriptive headers.
