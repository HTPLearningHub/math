# Histograms (Visualizing Distributions)

When analyzing quantitative data (numbers you can measure, like height, weight, or test scores), we often want to know how the numbers are spread out.
*   Are most students scoring around 80%?
*   Are there many students failing?
*   Is the data shaped like a "bell curve" (where most values are in the middle)?

A **Histogram** is a specific type of chart that groups numbers into intervals (called **Bins** or buckets) and shows how many numbers fall into each bin. It is one of the most important tools in statistics and data science because it shows you the "shape" of your data instantly.

### Simple Explanation with an Everyday Analogy
Imagine you are sorting a big basket of **apples** by their weight:
*   You set up 4 boxes (Bins):
    *   Box 1: 50g to 100g
    *   Box 2: 101g to 150g
    *   Box 3: 151g to 200g
    *   Box 4: 201g to 250g
*   You weigh each apple and drop it into the correct box.
*   At the end, you count how many apples are in each box. If Box 2 has the most apples, you know that most of your apples are medium-sized.
*   A histogram is just a bar graph that shows how many apples (frequency) fell into each weight range (bins).

Let's generate random student exam scores using NumPy's normal distribution. A **Normal Distribution** (or Gaussian distribution) is a statistical pattern where most data points cluster around a central average value (the mean), and the rest taper off symmetrically toward both extremes, forming a bell-shaped curve.

We will use `np.random.default_rng()` to create:
*   An average score of `80` (the center or **location** of our data).
*   A standard deviation of `10` (the **scale** or spread of our data).
*   We will use `np.clip()` to keep all scores strictly between `0` and `100`.

#### Creating a Basic Histogram

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Initialize our random generator
rng = np.random.default_rng(seed=42)

# 2. Generate 500 exam scores (normal distribution around 80)
raw_scores = rng.normal(loc=80, scale=10, size=500)

# 3. Clip scores so they are strictly within 0 and 100
scores = np.clip(raw_scores, 0, 100)

# 4. Create the histogram
# We specify bins=10 (divide data into 10 intervals)
# We set edgecolor="black" to see where each bar ends
plt.hist(scores, bins=10, color="skyblue", edgecolor="black")

# 5. Add details
plt.title("Exam Scores Distribution")
plt.xlabel("Scores")
plt.ylabel("Number of Students")

# We save the plot first (headless mode)
# plt.savefig("/workspace/scratch/basic_histogram.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
print("Basic histogram generated successfully!")
```

#### Customizing Bins

By default, Matplotlib calculates the bins automatically. However, you can pass a **list of numbers** to the `bins` parameter to define the exact boundaries of your intervals!

```python
import numpy as np
import matplotlib.pyplot as plt

scores = np.array([55, 62, 67, 71, 74, 78, 82, 85, 89, 91, 95, 98])

# Define custom boundaries: F (0-59), D (60-69), C (70-79), B (80-89), A (90-100)
custom_bins = [0, 60, 70, 80, 90, 100]

plt.hist(scores, bins=custom_bins, color="lightgreen", edgecolor="black")

plt.title("Grades Grouped by Boundaries")
plt.xlabel("Grade Intervals")
plt.ylabel("Number of Students")

# plt.savefig("/workspace/scratch/custom_bins_histogram.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
print("Custom bins histogram generated!")
```

### Common Mistakes Beginners Make
1.  **Confusing Bar Charts with Histograms**:
    *   Use a **Bar Chart** when comparing different categories (e.g., "Apples" vs. "Bananas"). There are spaces between the bars.
    *   Use a **Histogram** when visualizing continuous numerical ranges (e.g., "0g to 100g"). The bars are usually touching each other because the intervals are continuous.
2.  **Using too many or too few bins**:
    *   If you set `bins=2`, your data looks like two giant blocks (no detail).
    *   If you set `bins=100` with small datasets, you will have empty gaps between skinny bars. Try to stick between 5 to 20 bins for normal datasets.


#### Exercise 1 (Easy)
Generate an array of 100 random integers between `1` and `50` (inclusive) using NumPy. Create a histogram with `5` bins and set the edge color to `black`.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=123)
data = rng.integers(low=1, high=51, size=100)

plt.hist(data, bins=5, color="orange", edgecolor="black")
plt.title("Random Integers (1-50)")
plt.xlabel("Value Ranges")
plt.ylabel("Frequency")

# plt.savefig("/workspace/scratch/ex1_histogram.png")
plt.show()
plt.close()
```

#### Exercise 2 (Medium)
You are analyzing the heights of 1000 basketball players.
1. Generate heights using a normal distribution: mean = `195` cm, standard deviation = `8` cm.
2. Create a histogram with `15` bins, color it `"salmon"`, and add appropriate labels.


```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=99)
heights = rng.normal(loc=195, scale=8, size=1000)

plt.hist(heights, bins=15, color="salmon", edgecolor="black")
plt.title("Basketball Player Heights")
plt.xlabel("Height (cm)")
plt.ylabel("Count")

# plt.savefig("/workspace/scratch/ex2_histogram.png")
plt.show()
plt.close()
```
