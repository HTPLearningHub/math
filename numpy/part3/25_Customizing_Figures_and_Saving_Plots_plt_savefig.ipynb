# Customizing Figures and Saving Plots (`plt.savefig`)

When you create a chart, you often need to share it outside your Python program—such as inserting it into a PDF report, sharing it on Slack, or placing it in an Angular dashboard.

In this module, we will learn how to:
1.  Control the canvas size (`figsize`) and pixel density/sharpness (`dpi`).
2.  Export/Save our charts as highly optimized `.png` or `.pdf` files using **`plt.savefig()`**.

### Simple Explanation with an Everyday Analogy
*   **`figsize` (Figure Size)**: This is like choosing the size of your physical poster board (e.g., 8 inches by 5 inches).
*   **`dpi` (Dots Per Inch)**: This is like choosing how many ink dots the printer sprays on each inch. High DPI means a sharp, crisp picture. Low DPI means a blurry, pixelated image.
*   **`plt.savefig()`**: This is like taking a snapshot of your canvas and exporting it as a file on your computer hard drive.

### Code Examples

Let's look at how to control the figure properties and save it safely.

#### Setting Size and Sharpness
To set size and DPI, pass `figsize=(width, height)` (in inches) and `dpi` to `plt.figure()`.

````python
import numpy as np
import matplotlib.pyplot as plt

# Create a figure that is 8 inches wide, 4 inches tall, with 100 DPI
plt.figure(figsize=(8, 4), dpi=100)

x = np.arange(1, 6)
y = x * 2

plt.plot(x, y, marker='o')
plt.title("Configured Figure Canvas")
````

#### Saving a Plot as a PNG File
To save a file, call `plt.savefig('filename.png')`.

````python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.plot([1, 2, 3], [10, 20, 30])
plt.title("My Saved Plot")

# Save the plot with tight bounds so labels do not get cut off!
plt.savefig('my_saved_plot.png', dpi=300, bbox_inches='tight')
plt.close() # Safely close the figure to save system memory
````

### Common Mistakes Beginners Make
1.  **Calling `plt.savefig()` AFTER `plt.show()`**:
    ```python
    # ❌ WRONG WAY
    plt.plot(x, y)
    plt.show()
    plt.savefig('plot.png') # This will save a completely blank white image!
    ```
    *Why?* Because `plt.show()` automatically clears the current active figure canvas once it finishes rendering. Always call `plt.savefig()` **before** `plt.show()`!
2.  **Not using `bbox_inches='tight'`**: When saving plots, sometimes the axis labels or titles are close to the edge and get cropped (cut off). Adding `bbox_inches='tight'` automatically resizes the margins so everything fits perfectly in the saved file.


#### Exercise 1 (Easy)
Create a simple line plot and save it as a high-quality PDF named `'my_report_chart.pdf'` with `dpi=300`. Ensure no labels are cut off.


````python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.plot([1, 2, 3], [5, 15, 10])
plt.title("PDF Export Plot")

# We can save directly as PDF by changing the file extension!
plt.savefig('my_report_chart.pdf', dpi=300, bbox_inches='tight')
plt.close()
````
