# 3D Plotting with Matplotlib (`mplot3d`)

Most data we visualize is 2-dimensional (X and Y). But in advanced physics, geology, machine learning, or financial modeling, we often have three variables (X, Y, and Z) that interact.

Matplotlib includes a built-in toolkit called `mplot3d` that allows us to render 3-dimensional lines, scatter points, and surface terrains. Learning this allows you to create interactive, rotatable 3D visualizations.

### Simple Explanation with an Everyday Analogy
Imagine a **topographical mountain map**:
*   **2D Map**: Shows the latitude (X) and longitude (Y). You cannot easily see how high the mountain is.
*   **3D Map**: Adds elevation (Z - height). You can see the actual peaks and valleys of the mountain rising up from the ground.

### Code Examples

Let's see how to set up a 3D canvas and draw 3D lines and points.

#### Creating a 3D Scatter Plot
To create a 3D plot, we must add `projection='3d'` to our axes.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create standard figure
fig = plt.figure(figsize=(8, 6))

# Add a subplot with 3D projection
ax = fig.add_subplot(projection='3d')

# Generate 3D data points
z = np.linspace(0, 15, 100)
x = np.sin(z)
y = np.cos(z)

# Draw a 3D Scatter Plot
ax.scatter3D(x, y, z, c=z, cmap='viridis', s=50)

ax.set_title("3D Scatter Plot")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis") # Notice we have a Z label now!
```

#### Creating a 3D Line Plot
We can also connect our 3D points with a continuous line using `plot3D`.

```python
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(projection='3d')

z = np.linspace(0, 15, 200)
x = np.sin(z)
y = np.cos(z)

# Draw a continuous 3D helix (spiral)
ax.plot3D(x, y, z, 'red', linewidth=3)

ax.set_title("3D Line Plot (Helix)")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
```

###  Common Mistakes Beginners Make
1.  **Forgetting `projection='3d'`**: Writing `ax = fig.add_subplot()` without specifying `projection='3d'`. If you forget this, Matplotlib will try to render your 3D points on a flat 2D grid, resulting in a `ValueError` or a broken plot.
2.  **Mismatched Shapes for Z**: Your X, Y, and Z arrays must have the **exact same length/shape** so that each coordinate $(x_i, y_i, z_i)$ forms a valid point in space.


#### Exercise 1 (Easy)
Write a Python script to create a 3D scatter plot of 5 random points using NumPy's random number generator. Label all three axes.

```python
import numpy as np
import matplotlib.pyplot as plt

# Modern random generator
rng = np.random.default_rng(seed=42)

x = rng.uniform(0, 10, 5)
y = rng.uniform(0, 10, 5)
z = rng.uniform(0, 10, 5)

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.scatter3D(x, y, z, color='purple', s=100)

ax.set_xlabel("X Coordinate")
ax.set_ylabel("Y Coordinate")
ax.set_zlabel("Z Coordinate")
ax.set_title("5 Random 3D Points")
```

#### Exercise 2 (Medium)
Create a 3D line plot representing a straight diagonal line going from coordinate $(0,0,0)$ to $(10,10,10)$. Make the line blue and thick.


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate points along the diagonal
points = np.linspace(0, 10, 100)
x = points
y = points
z = points

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot3D(x, y, z, color='blue', linewidth=4)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("3D Diagonal Line")
```
