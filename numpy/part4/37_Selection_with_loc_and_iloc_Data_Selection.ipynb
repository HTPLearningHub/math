# Selection with `.loc` and `.iloc` (Data Selection)

During data analysis, we rarely need the entire table at once. Usually, we want to select specific rows, columns, or individual cells.
In Pandas, two main methods are used for selecting data: **`.loc`** and **`.iloc`**.
*   **`.loc` (Label-based)**: Performs data selection based on **labels**, using row and column names.
*   **`.iloc` (Integer position-based)**: Performs data selection strictly based on **integer positions (index positions)**, which always start from 0.

Understanding the difference between these two methods is a cornerstone of Pandas.

### Simple Explanation and Real-Life Analogy
Imagine a **contacts application (phonebook)**:
*   **`.loc` (Label-based)**: You search by name. For example, you say, "Find me *Anna's* phone number." It doesn't matter which row Anna's name is written in; you search specifically by that label.
*   **`.iloc` (Integer position-based)**: You search by position in the list. For example, you say, "Show me the very *first* contact in the phonebook" (position 0) or "the last contact." You do not specify the name, only the position number.

### Code Examples

First, let's create a simple DataFrame with programmer data, where we set their names as the index.

````python
import pandas as pd

data = {
    'Language': ['TypeScript', 'Python', 'JavaScript', 'Python', 'TypeScript'],
    'Experience': [3, 5, 2, 8, 1],
    'Salary': [85000, 110000, 75000, 145000, 60000]
}

df = pd.DataFrame(data, index=['Alex', 'John', 'Sarah', 'Emily', 'David'])
print(df)
````

#### Single Row Selection
*   Using **`.loc`**, we select by name (label).
*   Using **`.iloc`**, we select by integer position (starting from 0).

````python
# Selecting John's data (by label)
print("--- John's data (.loc) ---")
print(df.loc['John'])

# Selecting the second row (index 1, which is also John)
print("\n--- Row 1 data (.iloc) ---")
print(df.iloc[1])
````

#### Row & Column Selection (Getting a specific cell value)
In both methods, the syntax is `[row, column]`.

````python
# Using .loc (John's salary)
salary_loc = df.loc['John', 'Salary']
print("John's Salary (.loc):", salary_loc)

# Using .iloc (Row 1, Column 2 - Salary)
salary_iloc = df.iloc[1, 2]
print("John's Salary (.iloc):", salary_iloc)
````

#### Slicing with `.loc` vs `.iloc` (Data Slicing)
**ATTENTION (IMPORTANT DIFFERENCE):**
*   When slicing with `.loc`, **the endpoint is included (inclusive)**.
*   When slicing with `.iloc`, **the endpoint is NOT included (exclusive)**, just like standard slicing in Python lists or TypeScript arrays.


````python
# For .loc, 'Sarah' is included in the result
print("--- .loc slicing (from 'John' to 'Sarah') ---")
print(df.loc['John':'Sarah'])

# For .iloc, the row with index 3 ('Emily') is NOT included
print("\n--- .iloc slicing (from 1 to 3) ---")
print(df.iloc[1:3])
````

### 4. Common Mistakes Made by Beginners
1.  **Endpoint confusion during slicing**: Writing `df.iloc[1:3]` expecting to get rows 1, 2, and 3. But `.iloc` is exclusive of the end index (it only returns rows 1 and 2).
2.  **Incorrect brackets**: Writing `df.loc('John')` with regular parentheses. Remember, since this is indexing, you must always use **square brackets**: `df.loc['John']`.
3.  **Index confusion in tables with numeric index labels**: If your table's row labels are actual numbers (e.g., `0, 1, 2...`), `df.loc[0]` will look for the label named `0`, not necessarily the first row. To avoid this confusion, always use `.iloc` when you want to refer to absolute sequential positions.

#### Exercise 1 (Easy)
Create a Series object where the data represents the working hours of the week: `[8, 8, 8, 6, 5]`, and the indices are the names of the days: `['Mon', 'Tue', 'Wed', 'Thu', 'Fri']`.
1. Retrieve the hours of 'Thu' using `.loc`.
2. Retrieve the hours of the last day (index 4) using `.iloc`.

````python
import pandas as pd
hours = pd.Series([8, 8, 8, 6, 5], index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri'])

print("Hours on Thursday:", hours.loc['Thu'])
print("Hours on Friday:", hours.iloc[4])
````

#### Exercise 2 (Medium)
Given the following DataFrame:
```python
import pandas as pd
grid = pd.DataFrame([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
], index=['R1', 'R2', 'R3'], columns=['C1', 'C2', 'C3'])
```
Using `.loc` or `.iloc`, select the central 2x2 square (values: `50, 60` and `80, 90`).

````python
import pandas as pd
grid = pd.DataFrame([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
], index=['R1', 'R2', 'R3'], columns=['C1', 'C2', 'C3'])

# Using .loc (rows R2 to R3, columns C2 to C3)
print(grid.loc['R2':'R3', 'C2':'C3'])

print("----------------------------------")
# Or using .iloc (rows 1 to end, columns 1 to end)
print(grid.iloc[1:3, 1:3])
````
