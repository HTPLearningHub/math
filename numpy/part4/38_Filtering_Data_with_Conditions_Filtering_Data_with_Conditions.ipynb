# Filtering Data with Conditions (Filtering Data with Conditions)

In real-world data analysis, we often need to filter a table based on a specific condition. For example, "Show only those programmers whose salary is greater than $100,000".

In Pandas, filtering is done using the **Boolean Masking** technique. When we apply a comparison operator to a column of a DataFrame, Pandas returns a new Series containing `True` and `False` values. Then, we pass that Series to the DataFrame, and Pandas keeps only the rows corresponding to `True`.

### 2. Simple Explanation and Real-Life Analogy
Imagine that you are a **security guard** at the entrance of a nightclub:
*   You have a guest list. Next to each person, you stamp a mark: `True` (if they are older than 18) or `False` (if they are younger).
*   Then, you only let in those who have the `True` stamp.
*   This is exactly what filtering is: we create a True/False mask and clean our data using it.

### Code Examples

Let's use our previous programmers DataFrame.

#### Single Condition Filtering
Let's find the programmers who have more than 4 years of experience (`Experience > 4`).



````python
import pandas as pd

data = {
    'Language': ['TypeScript', 'Python', 'JavaScript', 'Python', 'TypeScript'],
    'Experience': [3, 5, 2, 8, 1],
    'Salary': [85000, 110000, 75000, 145000, 60000]
}

df = pd.DataFrame(data, index=['Alex', 'John', 'Sarah', 'Emily', 'David'])

# Step 1: Create the Boolean Mask
mask = df['Experience'] > 4
print("--- Boolean Mask ---")
print(mask)

# Step 2: Pass the mask to the DataFrame
filtered_df = df[mask]
print("\n--- Filtered Table ---")
print(filtered_df)
````

#### B) Multiple Conditions Filtering
Often we need to combine several conditions. In Pandas, special bitwise operators are used for this:
*   `&` corresponds to the **AND** condition.
*   `|` corresponds to the **OR** condition.
*   `~` corresponds to the **NOT (negation)** condition.

**VERY IMPORTANT RULES:**
1.  Each condition **MUST** be enclosed in parentheses: `(condition1) & (condition2)`.
2.  Standard Python keywords `and`, `or`, and `not` **DO NOT work** for filtering Pandas arrays.

Let's find the programmers who write **Python** **AND** have a salary **above $100,000**:

````python
# Filtering with multiple conditions (AND)
python_high_salary = df[(df['Language'] == 'Python') & (df['Salary'] > 100000)]
print(python_high_salary)
````

### 4. Common Mistakes Made by Beginners
1.  **Forgetting parentheses**: Writing `df[df['Salary'] > 80000 & df['Language'] == 'Python']`. This will immediately throw an error (ValueError/TypeError). **Always write with parentheses:** `df[(df['Salary'] > 80000) & (df['Language'] == 'Python')]`.
2.  **Using `and` and `or` words**: Trying to write `df[(df['Salary'] > 80000) and (df['Language'] == 'Python')]`. This will not work because `and` is only designed to evaluate single boolean values, whereas Pandas needs to compare entire columns element-wise.



#### Exercise 1 (Medium)
Given the following DataFrame with store products:
```python
import pandas as pd
products = pd.DataFrame({
    'Item': ['Laptop', 'Phone', 'Tablet', 'Headphones', 'Smartwatch'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories', 'Accessories'],
    'Price': [1200, 800, 450, 150, 250],
    'InStock': [True, False, True, True, False]
})
```
Write Pandas code that will filter and display only those products that are **in stock (InStock == True)** **AND** have a price **under $300**.


````python
import pandas as pd
products = pd.DataFrame({
    'Item': ['Laptop', 'Phone', 'Tablet', 'Headphones', 'Smartwatch'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories', 'Accessories'],
    'Price': [1200, 800, 450, 150, 250],
    'InStock': [True, False, True, True, False]
})

# Apply both conditions using parentheses and the & operator
filtered_products = products[(products['InStock'] == True) & (products['Price'] < 300)]
print(filtered_products)
````

*Explanation:* Only `Headphones` match both conditions. `Tablet` is in stock, but its price is over $300 ($450). `Smartwatch` is under $300, but it is not in stock.
