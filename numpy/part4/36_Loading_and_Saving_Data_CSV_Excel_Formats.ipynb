# Loading and Saving Data (CSV, Excel, & Formats)

In real-world development, we do not type out raw dictionaries of data. Instead, we load our data from external files (like CSVs, Excel spreadsheets, or Databases).

We will learn how to:
1. Load data using `pd.read_csv()` and `pd.read_excel()`.
2. Export data back using `.to_csv()` and `.to_excel()`.
3. Understand why file formats matter for **speed** and **size**.

* **Why it matters:** Excel has a hard limit of around **1 million rows** and gets very slow with large files.
* Pandas can easily process millions of rows and supports optimized modern formats like **Parquet** (very popular in Big Data because it is compressed and super fast to load).

### Simple Explanation with an Everyday Analogy
Imagine importing cargo:
* **CSV (Comma Separated Values)** is like a flat, plain wooden box. It is simple and readable by anyone, but it takes up a lot of physical space (large file size).
* **Excel (.xlsx)** is like a heavy decorated cabinet. It is nice because it has sheets and colors, but it is very slow to open and heavy to move.
* **Parquet** is like a high-tech vacuum-sealed storage box. It compresses the data to be tiny and opens instantly using specialized machine tools.

### Code Examples

Let's create a small dummy CSV file first to practice reading and writing.


````python
import pandas as pd

# 1. Let's create a DataFrame
devs = pd.DataFrame({
    "Name": ["Anna", "David"],
    "Role": ["Frontend", "Backend"]
})

# 2. Saving to a CSV file
# We use index=False so Pandas does not save the extra 0, 1 numbers column!
devs.to_csv("developers.csv", index=False)
print("File successfully saved!")
````

#### A) Loading the CSV File
We use `pd.read_csv()` to load it back:

````python
import pandas as pd

# Loading the CSV
loaded_df = pd.read_csv("developers.csv")
print("Loaded DataFrame:", loaded_df)
````

#### B) Loading Data directly from a URL (Web Scraping/Remote Fetch)
One of the coolest features of Pandas is that you can pass a **web URL** directly into `pd.read_csv()`! It downloads and loads the file automatically.

**But careful — the URL must point to the *raw* file, not to a web page.**

When you open a CSV on GitHub in your browser, the address looks like this:

```
https://github.com/USER/REPO/blob/master/folder/file.csv
```

That address gives you an **HTML page** (the pretty GitHub page with buttons, menus and colours) — not the plain numbers. Pandas downloads that HTML, tries to read it as a table of commas, and gets confused. That is why it says:

```
ParserError: Error tokenizing data. C error: Expected 1 fields in line 40, saw 2
```

*Tokenizing* just means "cutting the text into small pieces". Pandas cut line 40 of the HTML and found 2 pieces where it expected 1, so it stopped.

The fix: use the **raw** address. Change `github.com` to `raw.githubusercontent.com` and delete the word `blob`:

```
https://raw.githubusercontent.com/USER/REPO/master/folder/file.csv
```

(On the GitHub page there is also a **Raw** button that gives you exactly this link.)

````python
import pandas as pd

# The RAW address of the file: plain text, no web page around it.
# github.com -> raw.githubusercontent.com, and the word 'blob' is removed.
url = "https://raw.githubusercontent.com/KeithGalli/complete-pandas-tutorial/master/warmup-data/coffee.csv"

# read_csv downloads the text from the internet and builds a DataFrame from it
coffee_df = pd.read_csv(url)

# head(3) shows only the first 3 rows, so the output stays short
print(coffee_df.head(3))

# .shape tells us (number of rows, number of columns)
print("shape:", coffee_df.shape)
````

### 4. Common Mistakes Beginners Make
1. **Forgetting `index=False` when saving**: By default, `.to_csv()` saves the row numbers (`0, 1, 2...`) as a column. If you load and save the file multiple times, you will get ugly columns like `"Unnamed: 0"` repeated . Always use `index=False`!
2. **Missing third-party libraries for Excel**: Reading Excel files (`.xlsx`) requires a Python package called `openpyxl`. If you get a `ModuleNotFoundError` when calling `pd.read_excel()`, you need to run `pip install openpyxl` in your environment.

Write a Python script that:
1. Loads the coffee sales data from this URL: `https://raw.githubusercontent.com/KeithGalli/pandas-2025/master/warmup-data/coffee.csv`.
