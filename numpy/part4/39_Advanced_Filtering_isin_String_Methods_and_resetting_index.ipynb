# Advanced Filtering: `.isin()`, String Methods, and resetting index

Sometimes simple comparison operators (`>`, `<`, `==`) are not enough for complex searches. For example, if we need to filter data based on several possible values, or perform a search with a partial text match (like SQL's `LIKE` operator).

In this module, we will learn how to apply the **`.isin()`** method, text functions (via the **`.str`** attribute), and see how to properly reset table indices after filtering.

### 2. Simple Explanation and Real-Life Analogy
*   **`.isin()`**: Imagine you are shopping and have a list of your favorite brands: `['Apple', 'Samsung']`. You look at the store items and ask, "Is this item's brand **on my list**?" This is exactly the `.isin()` method.
*   **`.str.contains()`**: Imagine you are looking for books in a library that have the word "Python" in their title. It doesn't matter if the title starts or ends with it, as long as the text contains that word.

#### A) Filtering by a List of Values (`.isin()`)
If we want to select only programmers writing in 'TypeScript' or 'JavaScript', instead of writing multiple `==` conditions, we use `.isin()`.


````python
import pandas as pd
df = pd.DataFrame({
    'Language': ['TypeScript', 'Python', 'JavaScript', 'Python', 'TypeScript'],
    'Salary': [85000, 110000, 75000, 145000, 60000]
}, index=['Alex', 'John', 'Sarah', 'Emily', 'David'])

# Filter using .isin()
web_devs = df[df['Language'].isin(['TypeScript', 'JavaScript'])]
print(web_devs)
````

#### B) Text Search (`.str.contains()`)
To apply text-based functions on string columns in Pandas, you must first access the **`.str`** attribute.
Let's find the languages that contain the substring "Script".

````python
# Searching with partial text matching
script_languages = df[df['Language'].str.contains('Script')]
print(script_languages)
````

*Useful tip:* If you want the search to be case-insensitive, you can add the `case=False` parameter: `df['Language'].str.contains('script', case=False)`.

#### Index Resetting (`.reset_index()`)
When we filter a table, the original row index labels are preserved, which can leave gaps in the numbering (for example, row indices might look like `0, 2, 4...`).
To renumber the table sequentially from 0, 1, 2... we use the **`.reset_index(drop=True)`** method.


````python
# Create a DataFrame with standard numeric index
data = pd.DataFrame({
    'Name': ['Alex', 'John', 'Sarah', 'Emily'],
    'Age': [25, 30, 22, 35]
})

# Filter for people under 30
young_people = data[data['Age'] < 30]
print("--- After Filtering (Indices are non-sequential) ---")
print(young_people)

# Resetting the index (drop=True removes the old index column)
young_people_clean = young_people.reset_index(drop=True)
print("\n--- After Resetting Index ---")
print(young_people_clean)
````

### Common Mistakes Made by Beginners
1.  **Applying `.str` on non-string columns**: If you try to write `df['Salary'].str.contains(...)`, Pandas will throw an error (AttributeError) because `Salary` is a numeric column, not a string column.
2.  **Forgetting `drop=True` during `reset_index()`**: If you don't write `drop=True`, Pandas will turn your old non-sequential indices into a standard column (named `index`) and add new 0, 1, 2... sequential indices. Usually, we don't need the old index, so always write `drop=True`.

#### Exercise 1 (Medium)
Given the following database of employees:
```python
import pandas as pd
employees = pd.DataFrame({
    'Name': ['Michael Scott', 'Dwight Schrute', 'Jim Halpert', 'Pam Beesly', 'Angela Martin'],
    'Department': ['Sales', 'Sales', 'Sales', 'Admin', 'Accounting'],
    'Role': ['Manager', 'Assistant to the Regional Manager', 'Salesman', 'Receptionist', 'Head Accountant']
})
```
1. Filter employees working in the **Sales** or **Accounting** departments (use `.isin()`).
2. Filter employees whose job title (`Role`) contains the word **"Manager"** (use `.str.contains()`).

````python
import pandas as pd
employees = pd.DataFrame({
    'Name': ['Michael Scott', 'Dwight Schrute', 'Jim Halpert', 'Pam Beesly', 'Angela Martin'],
    'Department': ['Sales', 'Sales', 'Sales', 'Admin', 'Accounting'],
    'Role': ['Manager', 'Assistant to the Regional Manager', 'Salesman', 'Receptionist', 'Head Accountant']
})

# 1. Filtering by departments
dept_filter = employees[employees['Department'].isin(['Sales', 'Accounting'])]
print("--- Sales & Accounting ---")
print(dept_filter)

# 2. Filtering by role name
manager_filter = employees[employees['Role'].str.contains('Manager')]
print("\n--- Managers ---")
print(manager_filter)
````
