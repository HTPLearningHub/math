# The DataFrame Object (Tabular Data)

While a Series is a single column, a **DataFrame** is a full table with multiple rows and columns.

* **Why it matters:** In the real world, data is almost always multi-dimensional. For example, a table of developers has names, ages, programming languages, and salaries. A DataFrame lets us group all this different information together into one structured table.
* Since you are an Angular/TypeScript developer, you can think of a DataFrame as an **Array of Objects**. But Pandas stores it as a column-based table, which makes searching, sorting, and math extremely fast.

### 2. Simple Explanation with an Everyday Analogy
Imagine an **Excel spreadsheet** or a **Google Sheet**:
* The whole sheet is a **DataFrame**.
* Each single vertical column (like "Phone Number" or "Email") is a **Series**.
* The column headers at the top are the **column names**.
* The row numbers on the left are the **index**.

#### Creating a DataFrame from a Python Dictionary
The most common way to create a DataFrame from scratch is using a Python dictionary. The **keys** become the column names, and the **values** (lists) become the column data.



````python
import pandas as pd

# Dictionary containing dummy developer data
data = {
    "Name": ["Alex", "John", "Sarah", "Emily"],
    "Language": ["TypeScript", "TypeScript", "Python", "TypeScript"],
    "Salary": [80000, 95000, 110000, 85000]
}

df = pd.DataFrame(data)
print("Our DataFrame:", df)
````

#### B) Basic Inspection: Checking Your Data
Before doing any analysis, we must inspect our table. Pandas provides highly useful tools for this:
* `df.head(n)`: Shows the first `n` rows of the table.
* `df.tail(n)`: Shows the last `n` rows.
* `df.shape`: Returns a tuple showing `(rows, columns)`.
* `df.info()`: Displays columns, non-null counts, and data types.
* `df.describe()`: Provides quick statistical summaries (mean, min, max, etc.).

Let's test these methods:


````python
import pandas as pd

data = {
    "Name": ["Alex", "John", "Sarah", "Emily"],
    "Language": ["TypeScript", "TypeScript", "Python", "TypeScript"],
    "Salary": [80000, 95000, 110000, 85000]
}
df = pd.DataFrame(data)

# Check the shape (rows, columns)
print("DataFrame Shape:", df.shape)

# View the first 2 rows
print("First 2 rows:", df.head(2))

# View columns
print("Columns:", df.columns)
````

### 4. Common Mistakes Beginners Make
1. **Different list lengths in a dictionary**: When creating a DataFrame from a dictionary, **all lists must have the exact same length**. If one list has 4 items and another has 3, Pandas will throw an error.
2. **Capitalization in `pd.DataFrame`**: The `D` and the `F` **must be capitalized**. Writing `pd.dataframe` or `pd.Dataframe` will fail because it is a constructor class, not a simple function.

#### Exercise 3 (Hard)

Create a DataFrame of 5 employees with their names, hourly wages, and hours worked.
1. Add a new column called `"TotalPay"` which is calculated by multiplying hourly wage and hours worked.
2. Print the summary statistics of the numeric columns using `.describe()`.


````python
import pandas as pd

employees = {
    "Name": ["Amy", "Bob", "Charlie", "David", "Eva"],
    "Wage": [25.0, 30.0, 22.5, 40.0, 35.0],
    "Hours": [40, 35, 45, 38, 40]
}
df_emp = pd.DataFrame(employees)

# 1. Calculate TotalPay
df_emp["TotalPay"] = df_emp["Wage"] * df_emp["Hours"]

# 2. Print summary statistics
print(df_emp.describe())
````

*Explanation:* Pandas vectorizes the multiplication `df_emp["Wage"] * df_emp["Hours"]`. It automatically multiplies index-by-index in parallel and creates the `"TotalPay"` column.
