# Multi-Variable Visualization (Scatter Plots, Histograms, and Legends)

In this module, you will learn how to:
* Create a **Scatter Plot** to find correlations between two numerical variables (like HP and Attack).
* Create a **Histogram** to see the distribution of a single variable.
* Add custom **Legends** to distinguish different groups of data.

**Why it matters:** Real-world questions often require comparing multiple variables. For example, *"Do programmers with more experience earn higher salaries?"*. A scatter plot allows you to see this relationship instantly.

### Simple Explanation with an Everyday Analogy
Imagine you are a sports coach looking at your players:
* **Histogram**: You count how many players are between 150cm-160cm, 160cm-170cm, and 170cm-180cm. This shows the general spread of height in your team.
* **Scatter Plot**: You plot height on the X-axis and weight on the Y-axis. Each player is a dot. You can instantly see if taller players tend to weigh more (a positive correlation).

### Code Examples

Let's analyze the relationship between HP and Attack for different Pokémon types.

#### Plotting a Scatter Plot with Multiple Categories
We will plot HP on the X-axis and Attack on the Y-axis. We will color Fire Pokémon red and Water Pokémon blue.

```python
# pandas is the library for tables of data (rows and columns), like a spreadsheet.
# matplotlib is the library we use to draw charts.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Build the same small table we used in notebook 30, so this notebook can run on its own.
df = pd.DataFrame({
    'Name': ['Bulbasaur', 'Ivysaur', 'Venusaur', 'Charmander', 'Charmeleon',
             'Charizard', 'Squirtle', 'Wartortle', 'Blastoise', 'Caterpie'],
    'Type 1': ['Grass', 'Grass', 'Grass', 'Fire', 'Fire',
               'Fire', 'Water', 'Water', 'Water', 'Bug'],
    'Type 2': ['Poison', 'Poison', 'Poison', np.nan, np.nan,
               'Flying', np.nan, np.nan, np.nan, np.nan],
    'HP': [45, 60, 80, 39, 58, 78, 44, 59, 79, 45],
    'Attack': [49, 62, 82, 52, 64, 84, 48, 63, 83, 30],
})

# Same cleaning step as notebook 30: NaN means "missing value".
# .copy() makes a separate table so we do not change the original one.
df_clean = df.copy()
# .fillna('None') replaces every missing value with the text 'None'.
df_clean['Type 2'] = df_clean['Type 2'].fillna('None')

df_clean
```

```python
import matplotlib.pyplot as plt

# Filter Fire and Water Pokémon
fire_pokemon = df_clean[df_clean['Type 1'] == 'Fire']
water_pokemon = df_clean[df_clean['Type 1'] == 'Water']

plt.figure(figsize=(6, 4))

# Plot Fire Pokémon
plt.scatter(fire_pokemon['HP'], fire_pokemon['Attack'],
            color='red', label='Fire', s=150, alpha=0.8, edgecolor='black')

# Plot Water Pokémon
plt.scatter(water_pokemon['HP'], water_pokemon['Attack'],
            color='blue', label='Water', s=150, alpha=0.8, edgecolor='black')

plt.title("HP vs. Attack Correlation")
plt.xlabel("HP (Health Points)")
plt.ylabel("Attack Points")
plt.legend() # Show the Legend to identify types
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```


### Common Mistakes Beginners Make
1. **Forgetting `plt.legend()`**: If you use `label='...'` inside `plt.scatter()`, the labels will not show up on the image unless you explicitly call `plt.legend()`.
2. **Not Using `alpha` (Transparency)**: If you have many dots in the same spot, they will cover each other. Setting `alpha=0.6` makes dots semi-transparent, so you can see overlapping points.

#### Exercise 1 (Easy)
Create a NumPy array of 100 random values using a normal distribution centered at 70 with a standard deviation of 15.


```python
import numpy as np

# Set the seed for reproducibility
rng = np.random.default_rng(seed=42)

# Generate 100 values
values = rng.normal(loc=70, scale=15, size=100)
print("First 5 values:", values[:5])
```

#### Exercise 2 (Medium)
Plot a histogram of the random values from Exercise 1 using 10 bins and a black edge color.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 3))
# Plot histogram with 10 bins
plt.hist(values, bins=10, color='lightgreen', edgecolor='black')
plt.title("Distribution of Random Values")
plt.xlabel("Value Range")
plt.ylabel("Frequency")
plt.show()
```
