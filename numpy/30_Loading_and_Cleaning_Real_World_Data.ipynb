# Loading and Cleaning Real-World Data

When you work with real-world data, the data is almost never clean. It often has missing numbers, incorrect formats, or unnecessary columns. 

In this module, you will learn how to:
* Load a dataset from a CSV file.
* Inspect its shape and dimensions.
* Find and handle missing values, which are represented as **NaN** (Not a Number) in NumPy and Pandas.

**Why it matters:** If you try to run mathematical operations (like calculating the average) on an array containing missing values, the output might be `NaN`. Cleaning your data first is 80% of a data scientist's job!

### Simple Explanation with an Everyday Analogy
Imagine you are cooking a meal with fresh vegetables from a garden:
* **Loading**: Bringing the basket of vegetables into your kitchen.
* **Cleaning**: Washing off the dirt and cutting away the rotten parts. If you cook the vegetables with the dirt, the meal will taste bad.
* **In Data**: The dirt is **NaN** (missing values). We must find and clean these missing values before doing any calculations.

### Code Examples

Let's look at how to load a dataset and check for missing values using both Pandas and NumPy.

#### Loading and Inspecting the Dataset Shape
We can load a CSV file using Pandas [37]. Let's check how many rows and columns we have.


```python
import pandas as pd
import numpy as np
import os 

# Load the dataset
# 'data.csv' contains stats for 150 Pokémon
df = pd.read_csv('data.csv') if os.path.exists('data.csv') else pd.DataFrame({
    'Name': ['Bulbasaur', 'Ivysaur', 'Venusaur', 'Charmander', 'Charmeleon', 'Charizard', 'Squirtle', 'Wartortle', 'Blastoise', 'Caterpie'],
    'Type 1': ['Grass', 'Grass', 'Grass', 'Fire', 'Fire', 'Fire', 'Water', 'Water', 'Water', 'Bug'],
    'Type 2': ['Poison', 'Poison', 'Poison', np.nan, np.nan, 'Flying', np.nan, np.nan, np.nan, np.nan],
    'HP': [45, 60, 80, 39, 58, 78, 44, 59, 79, 45],
    'Attack': [49, 62, 82, 52, 64, 84, 48, 63, 83, 30]
})

print("Dataset Shape (Rows, Columns):", df.shape)
```

#### B) Finding and Handling NaN (Not a Number) Values
**NaN** is a special floating-point value used to represent missing data. Let's find out which columns have missing values.

```python
# Check which values are NaN in the 'Type 2' column
print("Is NaN Mask:\n", df['Type 2'].isna().values)

# Let's count how many missing values we have
missing_count = df['Type 2'].isna().sum()
print(f"\nNumber of missing values in 'Type 2': {missing_count}")
```

#### C) Filling NaN Values
We can fill missing values with a default string (like "None") so that our data is uniform and clean.

```python

# Fill NaN values in the 'Type 2' column with 'None'
df_clean = df.copy()
df_clean['Type 2'] = df_clean['Type 2'].fillna('None')

print("Cleaned 'Type 2' Column:\n", df_clean['Type 2'].values)
```

### 4. Common Mistakes Beginners Make
1. **Doing Math on Arrays with NaN**: If you use `np.mean()` on a NumPy array that contains `np.nan`, the result will be `nan`. Always clean or filter out NaNs first!
2. **Forgetting that Slices and Copies are Different**: If you modify `df['Type 2']` directly, you might get a warning. Always use `.copy()` when making a separate, clean DataFrame.

#### Exercise 1 (Easy)
Create a 1D NumPy array with values `[10.0, np.nan, 30.0, 40.0, np.nan]`. Use a NumPy function to check which elements are NaN.

```python
import numpy as np

arr = np.array([10.0, np.nan, 30.0, 40.0, np.nan])
# Use np.isnan() to find NaN values
nan_mask = np.isnan(arr)
print("NaN Mask:", nan_mask)
```

#### Exercise 2 (Medium)
Given the array from Exercise 1, replace all NaN values with `0.0`.


```python
import numpy as np

arr = np.array([10.0, np.nan, 30.0, 40.0, np.nan])
# Replace NaN values using the boolean mask
arr[np.isnan(arr)] = 0.0
print("Cleaned Array:", arr)
```
