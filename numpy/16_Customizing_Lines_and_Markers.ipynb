# Customizing Lines and Markers

By default, Matplotlib draws a simple, thin blue line. But in the real world, you want your charts to look professional and be easy to read.

In this module, you will learn how to customize:
*   **Markers**: Symbols (like circles, stars, or squares) placed at each data point to make the exact values clear.
*   **Line Styles**: Solid, dashed, or dotted lines.
*   **Colors**: Custom professional hex colors.
*   **Style Dictionaries**: How to store your custom styles in a Python dictionary and reuse them for multiple lines easily to avoid duplicate code.

### 2. Simple Explanation with an Everyday Analogy
Think of a **map with hiking trails**:
*   The path is a **dashed line** so you know it is a trail, not a paved road.
*   Each campsite or landmark along the way has a **star symbol**.
*   The line is colored **green** to match the forest theme.

Markers and line styles are the "design theme" of your chart!

#### Technical Term Explainers:
*   **Marker**: The shape used to represent data points (e.g., `o` for circle, `*` for star).
*   **Hex Color**: A code starting with `#` followed by 6 letters/numbers that represents a highly specific color (e.g., `#FF5733` is vibrant orange).
*   **Dictionary Unpacking**: Using double asterisks `**my_dict` in Python to pass a dictionary of settings directly into a function.




#### Adding Markers, Colors, and Custom Line Styles
Let's make a beautiful custom plot!

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4])
y = np.array([10, 25, 18, 35])

# Customize the plot using keyword arguments:
# - marker: 'o' for circles [6]
# - markersize: size of the circle (shortcut: ms) [7]
# - markerfacecolor: inner color of the marker (shortcut: mfc) [7]
# - markeredgecolor: outer border color of the marker (shortcut: mec) [8]
# - linestyle: 'dashed' (or '--'), 'dotted' (or ':') [8]
# - linewidth: thickness of the line (shortcut: lw) [8]
# - color: hex color for the line [9]
plt.plot(x, y,
         marker='o',
         markersize=12,
         markerfacecolor='#FFC300', # Warm Yellow [7]
         markeredgecolor='#900C3F', # Dark Red Border [8]
         linestyle='dashed', # Dashed line style [8]
         linewidth=3,
         color='#900C3F') # Match line color to border [9]

plt.show()
```

#### B) Plotting Multiple Lines & Using Style Dictionaries
If you are plotting multiple lines, you can pass them one by one. To keep your code clean, you can put all your styles into a dictionary and unpack it!

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4])
y1 = np.array([10, 20, 15, 30]) # Line 1 [9]
y2 = np.array([5, 15, 25, 12])  # Line 2 [9]

# Put common styles in a dictionary! [10]
my_style = {
    'marker': '*',       # Star marker [6]
    'markersize': 10,    # [7]
    'linestyle': ':',    # Dotted line [8]
    'linewidth': 2       # [8]
}

# Plot both lines. Unpack the dictionary using ** [10]
# We can also pass different colors as individual arguments [11]
plt.plot(x, y1, color='#1F77B4', **my_style) # Blue line [11]
plt.plot(x, y2, color='#2CA02C', **my_style) # Green line [11]

plt.show()
```


### 4. Common Mistakes Beginners Make
1.  **Confusing `color` with `markerfacecolor`**: Setting `color='blue'` and wondering why the inner circles are still red. `color` controls the **line**, while `markerfacecolor` (or `mfc`) controls the **inside of the marker**.
2.  **Using invalid marker symbols**: Writing `marker='square'` instead of `marker='s'`, or `marker='circle'` instead of `marker='o'`. Matplotlib uses short characters for marker shapes.
3.  **Forgetting the double asterisks `**` when unpacking a dictionary**: Writing `plt.plot(x, y, my_style)` instead of `plt.plot(x, y, **my_style)`. Without `**`, Matplotlib will think you are trying to pass the dictionary as coordinates and crash!

#### Exercise 1
Create a NumPy array with values `[2, 4, 6, 8]`. Plot these values with a **dotted line**, a line width of **5**, and **star markers** of size **15**.


```python
import numpy as np
import matplotlib.pyplot as plt

y = np.array([2, 4, 6, 8])
x = np.arange(len(y)) # [0, 1, 2, 3] as default X [5]

plt.plot(x, y, linestyle=':', linewidth=5, marker='*', markersize=15) # [6, 8]
plt.show()
```

#### Exercise 2
Given these two datasets:
*   `y1 = np.array([100, 120, 110, 130])`
*   `y2 = np.array([90, 95, 105, 125])`
Create a dictionary named `custom_line_style` with a solid line style, width of 4, and circle markers. Plot both datasets on the same chart using different colors.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4])
y1 = np.array([100, 120, 110, 130])
y2 = np.array([90, 95, 105, 125])

custom_line_style = {
    'marker': 'o',
    'linestyle': 'solid',
    'linewidth': 4
}

# Plot both lines by unpacking the dictionary
plt.plot(x, y1, color='blue', **custom_line_style)
plt.plot(x, y2, color='orange', **custom_line_style)

plt.show()
```
