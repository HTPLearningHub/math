# Scatter Plots (Finding Correlations)

A **scatter plot** displays individual points on a 2D plane. It does not connect the points with lines. Instead, it uses dot markers to show how much one variable is affected by another.

This is the ultimate tool for finding **correlations** (relationships) between variables:
*   **Positive Correlation**: As X increases, Y also increases (e.g., more hours of study leads to higher test scores).
*   **Negative Correlation**: As X increases, Y decreases (e.g., as the price of a service increases, the number of users decreases).
*   **No Correlation**: The dots are scattered randomly with no clear pattern (e.g., your shoe size and your typing speed have no relationship).

In engineering, we use scatter plots to map metrics, find data clusters, or identify anomalies.

### Simple Explanation with an Everyday Analogy
Imagine you are walking around a city and looking at different houses:
*   On the horizontal axis (X), you look at how big the house is.
*   On the vertical axis (Y), you look at how much the house costs.
*   If you put a single dot for every house, you will see a clear **positive correlation upward trend**: larger houses almost always cost more money.



#### Creating a Simple Scatter Plot
Let's plot study hours vs. exam scores for 11 students.

```python
import numpy as np
import matplotlib.pyplot as plt

# X-axis: Hours studied
hours_studied = np.array([0, 1, 1.5, 2, 3, 4, 5, 6, 7, 7.5, 8])
# Y-axis: Exam score (%)
exam_scores = np.array([55, 60, 62, 62, 68, 70, 75, 78, 82, 85, 87])

# Create the scatter plot
plt.scatter(hours_studied, exam_scores, color="darkblue", s=100, alpha=0.8)

plt.title("Study Hours vs. Exam Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score (%)")
plt.grid(True, linestyle="--", alpha=0.5)

# plt.savefig("basic_scatter_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```

*Formatting Parameters Explained:*
*   `s=100`: Changes the size of the circular dot markers to `100`. The default size is often too small for presentations.
*   `alpha=0.8`: Controls the **transparency** of the dots from `0.0` (completely invisible) to `1.0` (completely opaque). Setting a transparency value like `0.8` or `0.5` is extremely helpful when many dots lie on top of each other. It lets you see overlapping areas clearly!


#### Plotting Multiple Datasets with a Legend
Often, we want to compare two different groups (e.g., students from Class A and students from Class B) on the same graph. We can call `plt.scatter()` twice with different colors and labels, and then show a **legend** to identify them.


```python
import numpy as np
import matplotlib.pyplot as plt

# Dataset for Class A
hours_class_a = np.array([1, 2, 3, 5, 6, 8])
scores_class_a = np.array([60, 65, 68, 75, 78, 85])

# Dataset for Class B
hours_class_b = np.array([0, 1.5, 3.5, 4, 7, 7.5])
scores_class_b = np.array([50, 58, 70, 72, 83, 88])

# Plot Class A (blue circles)
plt.scatter(hours_class_a, scores_class_a, color="blue", s=150, label="Class A", alpha=0.7)

# Plot Class B (red circles)
plt.scatter(hours_class_b, scores_class_b, color="red", s=150, label="Class B", alpha=0.7)

# Add title, labels, and grid
plt.title("Study Hours comparison between Classes")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score (%)")
plt.grid(True, linestyle=":")

# Crucial: Show the legend to tell the reader which color represents which class
plt.legend()

# plt.savefig("comparison_scatter_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```

### 4. Common Mistakes Beginners Make
1.  **Forgetting to call `plt.legend()`**: If you pass `label="Class A"` inside your scatter function but forget to call `plt.legend()`, the legend block **will not appear** on your chart. Always add `plt.legend()`!
2.  **Using `plt.plot()` instead of `plt.scatter()`**: If you use `plt.plot(x, y)`, Matplotlib will draw a single zig-zag line connecting all the dots together. This looks chaotic and makes it impossible to see the statistical correlation. Always use `plt.scatter()` for scatter data.


#### Exercise 1 (Easy)
You are tracking your website's speed. You have a dataset of page weight (in Megabytes) vs. page load time (in seconds):
*   Page Weight (MB): `[1.2, 1.8, 2.5, 3.0, 3.8]`
*   Load Time (seconds): `[1.5, 2.1, 2.9, 3.2, 4.0]`

Create a scatter plot of this data. Customize it with red dots of size `150`. Give it titles and labels. Save the image as `page_load_scatter.png`.

```python
import numpy as np
import matplotlib.pyplot as plt

page_weight = np.array([1.2, 1.8, 2.5, 3.0, 3.8])
load_time = np.array([1.5, 2.1, 2.9, 3.2, 4.0])

plt.scatter(page_weight, load_time, color="red", s=150)
plt.title("Page Weight vs. Load Time")
plt.xlabel("Page Weight (MB)")
plt.ylabel("Load Time (seconds)")

# plt.savefig("page_load_scatter.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```

#### Exercise 2 (Medium)
You want to analyze if there is a relationship between how many cups of coffee a developer drinks in a day and the number of bugs they write:
*   Developer Coffee Cups: `[0, 1, 2, 2, 3, 4, 5, 6]`
*   Bugs Written: `[8, 6, 4, 5, 2, 1, 3, 5]`

Create a scatter plot. Enable the grid lines, set the dot transparency to `0.6`, and color the dots "darkgreen". Is there a correlation? Save the image as `coffee_bugs_scatter.png`.


```python
import numpy as np
import matplotlib.pyplot as plt

coffee = np.array([0, 1, 2, 2, 3, 4, 5, 6])
bugs = np.array([8, 6, 4, 5, 2, 1, 3, 5])

plt.scatter(coffee, bugs, color="darkgreen", s=180, alpha=0.6)
plt.title("Coffee Consumption vs. Bugs Written")
plt.xlabel("Cups of Coffee (Daily)")
plt.ylabel("Bugs Written")

# Add grid for easy coordinate lookup
plt.grid(True, linestyle="--", alpha=0.4)

# plt.savefig("coffee_bugs_scatter.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()

# Analytical note on correlation:
# Looking at the data, up to 4 cups of coffee, there is a clear negative correlation
# (more coffee leads to fewer bugs). However, after 4 cups (5 and 6),
# the number of bugs starts to increase again due to caffeine jitters!
```
