# Pie Charts (Visualizing Parts of a Whole)

A **pie chart** is a circular chart divided into slices. Each slice represents a percentage of the total amount.

In web dashboards, pie charts are used to show:
*   The **budget distribution** of a department.
*   The **operating system share** of your active web users (iOS vs. Android vs. Desktop).
*   The **status of tasks** in a project (Completed vs. In Progress vs. Blocked).

### Simple Explanation with an Everyday Analogy
Imagine a real **pizza** or a **pie**. If 4 friends order a pizza, and one person pays for half ($50\%$), that person gets exactly half of the pizza. If the other three friends split the remaining cost equally, they each get smaller, equal slices. The size of each slice represents their financial contribution.


#### Basic Pie Chart with Custom Labels, Percentages, and Shadows
Let's build a pie chart representing the enrollment numbers at a fictional college:
*   Freshman: 300 students
*   Sophomores: 250 students
*   Juniors: 275 students
*   Seniors: 225 students


```python
import numpy as np
import matplotlib.pyplot as plt

classes = np.array(["Freshman", "Sophomores", "Juniors", "Seniors"])
student_counts = np.array([300, 250, 275, 225])

# Define colors for each slice
slice_colors = ["#ff7f0e", "#2ca02c", "#d62728", "#9467bd"]

# Create the pie chart
plt.pie(
    student_counts,
    labels=classes,
    colors=slice_colors,
    autopct="%1.1f%%",  # Formats the percentages to 1 decimal place with a % symbol
    shadow=True,        # Adds a beautiful drop-shadow effect under the chart
    startangle=90       # Rotates the start of the pie chart by 90 degrees
)

plt.title("College Student Distribution")
# plt.savefig("basic_pie_chart.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```

*Key Parameters Explained:*
*   `autopct="%1.1f%%"`: This is a formatting code. `%1.1f` tells Matplotlib to calculate the percentages automatically and show exactly one decimal place. The double `%%` at the end prints a literal `%` character.
*   `shadow=True`: Adds a nice drop-shadow effect to make the pie chart look 3D and modern.
*   `startangle=90`: By default, Matplotlib starts drawing slices at 0 degrees (the right side of the circle). Setting `startangle=90` starts drawing from the top (12 o'clock), which is the standard layout for pie charts.


#### B) Making a Slice "Explode"
Sometimes, you want to draw the reader's attention to one specific slice (for example, the "Seniors" class). You can do this by using the **`explode`** parameter.

`explode` accepts an array of decimal numbers. Each number represents how far that specific slice should move outward from the center. A value of `0` means the slice stays in place, while a value of `0.2` moves it out slightly.


```python
import numpy as np
import matplotlib.pyplot as plt

classes = np.array(["Freshman", "Sophomores", "Juniors", "Seniors"])
student_counts = np.array([300, 250, 275, 225])

# We only want to explode the 4th slice (Seniors) by 0.15 units
explode_settings = [0, 0, 0, 0.15]

plt.pie(
    student_counts,
    labels=classes,
    autopct="%1.1f%%",
    explode=explode_settings,
    shadow=True,
    startangle=90
)

plt.title("College Student Distribution (Seniors Highlighted)")
# plt.savefig("exploded_pie_chart.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```

### 4. Common Mistakes Beginners Make
1.  **Forgetting to match the length of lists**: If you have 4 categories, your `student_counts`, `labels`, `colors`, and `explode` lists **must all have exactly 4 items**! If there is a length mismatch, Python will throw a `ValueError`.
2.  **Using too many slices**: If your dataset has 15 categories, your pie chart will have 15 tiny, unreadable slices. For datasets with more than 5 or 6 categories, a **bar chart** is almost always a better choice than a pie chart.


#### Exercise 1 (Easy)
You are monitoring your server's memory usage and discover that:
*   Operating System: uses 2.0 GB
*   Angular App: uses 1.5 GB
*   Database Cache: uses 4.5 GB

Create a pie chart with clear labels showing how your memory is distributed. Customize it with a drop-shadow and set the start angle to 90 degrees. Save the image as `server_memory_pie.png`.

```python
import numpy as np
import matplotlib.pyplot as plt

items = np.array(["Operating System", "Angular App", "Database Cache"])
memory_gb = np.array([2.0, 1.5, 4.5])

plt.pie(
    memory_gb,
    labels=items,
    autopct="%1.1f%%",
    shadow=True,
    startangle=90
)
plt.title("Server Memory Distribution (GB)")
# plt.savefig("server_memory_pie.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```

#### Exercise 2 (Medium)
You are managing a software team. You want to visualize the status of your current tasks:
*   `Backlog`: 40 tasks
*   `In Progress`: 25 tasks
*   `Under Review`: 15 tasks
*   `Done`: 80 tasks

Create a pie chart showing these task counts. Highlight the `Done` slice by "exploding" it by `0.1` units. Use the colors `gray` for Backlog, `blue` for In Progress, `orange` for Under Review, and `green` for Done. Set the percentages to show one digit after the decimal point. Save the image as `project_tasks_pie.png`.


```python
import numpy as np
import matplotlib.pyplot as plt

statuses = np.array(["Backlog", "In Progress", "Under Review", "Done"])
task_counts = np.array([40, 25, 15, 80])

# Backlog index 0, In Progress index 1, Under Review index 2, Done index 3
explode_settings = [0, 0, 0, 0.1]
status_colors = ["gray", "blue", "orange", "green"]

plt.pie(
    task_counts,
    labels=statuses,
    colors=status_colors,
    explode=explode_settings,
    autopct="%1.1f%%",
    shadow=True,
    startangle=90
)
plt.title("Project Task Status distribution")
# plt.savefig("project_tasks_pie.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```
