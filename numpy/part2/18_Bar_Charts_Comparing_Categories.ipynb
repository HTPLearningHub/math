# Bar Charts (Comparing Categories)

A **bar chart** compares different categories of data. It represents each category with a rectangular bar. The length or height of the bar is proportional to the value it represents.

In web development, you might use a bar chart to compare:
*   The **bundle size** of different Angular libraries in your project.
*   The **HTTP response time** of different API endpoints.
*   The **monthly page views** of different routes.

Matplotlib provides `plt.bar()` for vertical bar charts, and `plt.barh()` for horizontal bar charts.

### 2. Simple Explanation with an Everyday Analogy
Imagine you are packing bags with books:
*   The **vertical bar chart** is like placing books upright on a desk side-by-side. The tallest book represents the category with the highest value.
*   The **horizontal bar chart** is like stacking books flat on top of each other, extending to the right. The book that reaches the furthest to the right represents the largest value.



#### Creating a Vertical Bar Chart
Let's build a vertical bar chart to represent daily servings of different food categories.

```python
import numpy as np
import matplotlib.pyplot as plt

# We define the categories (X-axis) and daily servings consumed (Y-axis)
categories = np.array(["Grains", "Fruit", "Vegetables", "Protein", "Dairy", "Sweets"])
servings = np.array([4, 3, 2, 5, 3, 1])

# Create the vertical bar chart
plt.bar(categories, servings, color="skyblue")

# Add descriptive labels and title
plt.title("Daily Servings Consumed")
plt.xlabel("Food Category")
plt.ylabel("Servings (Quantity)")

# Save the plot as an image instead of using plt.show()
# plt.savefig("vertical_bar_chart.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```


#### Creating a Horizontal Bar Chart (`plt.barh()`)
Sometimes, your category names are very long (like full URLs or long department names). If you use a vertical bar chart, the names on the X-axis will overlap and look messy.
A **horizontal bar chart** is the perfect solution because it gives long names plenty of room on the left side of the chart.

To make it horizontal, we use `plt.barh()`:

```python
import numpy as np
import matplotlib.pyplot as plt

categories = np.array(["Grains", "Fruit", "Vegetables", "Protein", "Dairy", "Sweets"])
servings = np.array([4, 3, 2, 5, 3, 1])

# Create a horizontal bar chart
plt.barh(categories, servings, color="salmon")

plt.title("Daily Servings Consumed (Horizontal View)")
plt.xlabel("Servings (Quantity)")
plt.ylabel("Food Category")

# plt.savefig("horizontal_bar_chart.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```

### 4. Common Mistakes Beginners Make
1.  **Swapping the X and Y axes labels**: When you switch from a vertical bar chart to a horizontal bar chart (`plt.barh()`), the categories move to the Y-axis and the numerical values move to the X-axis. Remember to swap your `plt.xlabel()` and `plt.ylabel()` labels so they make sense!
2.  **Using too many categories**: If you try to display 50 different bars in a single chart, the bars will become extremely thin and impossible to read. If you have too many categories, filter out the smallest ones first or group them into an "Other" category.


#### Exercise 1 (Easy)
You want to compare the render times of three Angular components in your app:
*   `HeaderComponent`: 15 milliseconds (ms)
*   `DashboardComponent`: 120 ms
*   `FooterComponent`: 10 ms

Create a vertical bar chart with these components on the X-axis and their render times on the Y-axis. Set the color of the bars to "lightcoral". Save the image as `component_render_times.png`.


```python
import numpy as np
import matplotlib.pyplot as plt

components = np.array(["Header", "Dashboard", "Footer"])
render_times = np.array([15, 120, 10])

plt.bar(components, render_times, color="lightcoral")
plt.title("Angular Component Render Times")
plt.xlabel("Component Name")
plt.ylabel("Render Time (ms)")

plt.savefig("component_render_times.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```

#### Exercise 2
You have a dataset of browser market share:
*   `Chrome`: 65%
*   `Safari`: 18%
*   `Firefox`: 3%
*   `Edge`: 5%
*   `Other`: 9%

Create a horizontal bar chart representing this data. Customize it by setting the bar color to a custom hexadecimal color code, and add a black edge border to the bars (Hint: use the `edgecolor="black"` argument inside `plt.barh()`). Save the image as `browser_share.png`.


```python
import numpy as np
import matplotlib.pyplot as plt

browsers = np.array(["Chrome", "Safari", "Edge", "Firefox", "Other"])
shares = np.array([65, 18, 5, 3, 9])

# Sort the browsers so the most popular is at the top of our horizontal chart
# This makes it much easier to read!
plt.barh(browsers[::-1], shares[::-1], color="#3498db", edgecolor="black")

plt.title("Browser Market Share (%)")
plt.xlabel("Percentage Share")
plt.ylabel("Browser")

# plt.savefig("browser_share.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close()
```
