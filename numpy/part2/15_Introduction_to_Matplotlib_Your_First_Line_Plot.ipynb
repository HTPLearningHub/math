# Introduction to Matplotlib & Your First Line Plot

In this module, you will learn how to install Matplotlib, import its plotting module, and draw a simple **Line Plot**.

A **Line Plot** is a chart that displays information as a series of data points connected by straight line segments.
It is extremely useful for showing trends over time (for example, how website traffic or stock prices change month by month).

### Simple Explanation with an Everyday Analogy
Imagine you are drawing a graph on **graph paper**:
1. You draw a horizontal line at the bottom (the **X-axis**).
2. You draw a vertical line on the left side (the **Y-axis**).
3. You mark points on the paper where the X and Y values meet.
4. You take a ruler and connect the points with a line.

Matplotlib does exactly this, but in less than a millisecond!

Let's see how to write this in Python.

#### Technical Term Explainers:
*   **Module**: A Python file containing pre-written code you can import and use.
*   **Alias**: A nickname we give to an imported library to make our code shorter (like `import numpy as np`).
*   **Pyplot**: A sub-module inside Matplotlib that provides easy, user-friendly functions for drawing. We import it as `plt`.



#### Your First Basic Line Plot
Let's create a simple plot using standard Python lists first.

```python
import matplotlib.pyplot as plt # Import the plotting module

# X-coordinates (Years)
years = [2023, 2024, 2025, 2026]
# Y-coordinates (Class sizes / Number of students)
students = [15, 25, 30, 20]

# Draw the line plot
plt.plot(years, students)

# Display the plot window
# Note: In a headless environment or script, this saves a file.
# In interactive editors, this pops open a window!
plt.show()
```

#### B) Using NumPy Arrays with Matplotlib
Because NumPy arrays are faster and have much more math functionality, we should use them instead of standard lists. Matplotlib works perfectly with NumPy!

```python
import numpy as np
import matplotlib.pyplot as plt

# Create NumPy arrays
x_years = np.array([2023, 2024, 2025, 2026])
y_students = np.array([15, 25, 30, 20])

# Plot the arrays
plt.plot(x_years, y_students)
plt.show()
```

### 4. Common Mistakes Beginners Make
1.  **Forgetting `plt.show()`**: Writing `plt.plot(x, y)` and wondering why no chart appears on the screen. Always remember to call `plt.show()` to render and display the image!
2.  **Mismatching array sizes**: Passing an X array with 4 elements and a Y array with 5 elements. This will throw a `ValueError` because every point needs both an X and a Y coordinate.
3.  **Forgetting to import the `pyplot` sub-module**: Writing `import matplotlib` and then calling `matplotlib.plot()`. This will fail. You must import `matplotlib.pyplot`.

#### Exercise 1
Create a NumPy array for X with numbers from 1 to 5. Create a NumPy array for Y where each value is the square of X ($Y = X^2$). Plot this curve.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 6) # [1, 2, 3, 4, 5]
y = x ** 2 # [1, 4, 9, 16, 25] (Vectorized power!)

plt.plot(x, y)
plt.show()
```
