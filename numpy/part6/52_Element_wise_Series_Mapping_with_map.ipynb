# Element-wise Series Mapping with `.map()`

Often, data is stored in short codes or abbreviations to save space (e.g., country codes like `"USA"`, `"GBR"`, or department codes like `"MKT"`, `"ENG"`). When presenting data to business users, you need to "translate" these abbreviations into full, readable names.

The `.map()` method is designed specifically for **Series (columns)** to map each value from an old value to a new value. You can pass either:
*   A **Python dictionary** (where keys are the old values and values are the new ones).
*   A **Python function** or `lambda` to perform element-wise mapping.

If a value in the Series is not found in your mapping dictionary, Pandas will replace it with `NaN` (null value) by default. This makes `.map()` an excellent tool for strict categorical translations.

### Plain English Explanation & Real-World Analogy
Imagine you are at an international airport, and you have a list of flight codes like `["LHR", "JFK", "CDG"]`.
To find out where these flights are going, you look at a **translation guidebook** (a dictionary):
*   `"LHR"` $\rightarrow$ `"London Heathrow"`
*   `"JFK"` $\rightarrow$ `"John F. Kennedy (New York)"`
*   `"CDG"` $\rightarrow$ `"Charles de Gaulle (Paris)"`

Using `.map()` is like giving Pandas this guidebook and saying: *"Go through my entire column of airport codes and write down the full name of each city next to it."*

### Code Examples

Let's create a DataFrame representing employees with short department and country codes:


```python
import pandas as pd

data = {
    'Name': ['Liam', 'Olivia', 'Emma', 'Oliver', 'Sophia'],
    'Dept_Code': ['ENG', 'MKT', 'ENG', 'HR', 'MKT'],
    'Country_Code': ['USA', 'GBR', 'USA', 'USA', 'FRA']
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

#### Mapping with a Dictionary
Let's translate the department codes into full department names using a dictionary mapping:

```python
# Create the translation dictionary
dept_mapping = {
    'ENG': 'Engineering',
    'MKT': 'Marketing',
    'HR': 'Human Resources'
}

# Apply map to the column
df['Department'] = df['Dept_Code'].map(dept_mapping)
print(df[['Name', 'Dept_Code', 'Department']])
```


#### Handling Missing Keys (The NaN Pitfall)
If a value in the column is *not* present in the dictionary, it gets mapped to `NaN`. Let's see what happens to France (`FRA`) if our country dictionary only contains `USA` and `GBR`:

```python
country_mapping = {
    'USA': 'United States',
    'GBR': 'United Kingdom'
}

# Apply map - 'FRA' is missing from country_mapping
df['Country'] = df['Country_Code'].map(country_mapping)
print(df[['Name', 'Country_Code', 'Country']])
```

*Note:* To avoid losing unmapped values, you can use `.replace()` instead of `.map()`, or use `.fillna()` afterward to fill the nulls.

### 4. Common Mistakes Beginners Make
1.  **Trying to call `.map()` on an entire DataFrame**: `.map()` is a Series method. If you try to run `df.map(dictionary)`, it will fail or not work as expected in older Pandas versions. Only call it on a single column (Series): `df['column'].map(...)`.
2.  **Forgetting that unmapped values turn into `NaN`**: If your column has 10 unique values, but your dictionary only maps 8 of them, the remaining 2 will become `NaN` in the output! Be sure your dictionary covers all possible categories, or use `.replace()` if you want to keep unmapped values unchanged.

#### Exercise 1 (Easy)
Given a Series of student grades: `grades = pd.Series(['A', 'B', 'C', 'A', 'F'])`.
Use `.map()` to convert these letter grades into descriptive strings:
*   `'A'` $\rightarrow$ `'Excellent'`
*   `'B'` $\rightarrow$ `'Good'`
*   `'C'` $\rightarrow$ `'Average'`
*   `'F'` $\rightarrow$ `'Failed'`

```python
import pandas as pd

grades = pd.Series(['A', 'B', 'C', 'A', 'F'])

# Define mapping dictionary
grade_guide = {
    'A': 'Excellent',
    'B': 'Good',
    'C': 'Average',
    'F': 'Failed'
}

# Map letter grades to words
descriptive_grades = grades.map(grade_guide)
print(descriptive_grades)
```
