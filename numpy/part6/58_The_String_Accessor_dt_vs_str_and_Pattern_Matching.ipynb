# The String Accessor (`.dt` vs `.str`) & Pattern Matching

Pandas stores text columns as `object` or `string` data types. To perform string operations on every element of a column simultaneously without using slow Python `for` loops, you must use the **`.str` accessor**.

In this module, you will learn:
* How the `.str` accessor works under the hood.
* How to perform basic pattern matching using `.str.startswith()`, `.str.endswith()`, and `.str.contains()`.
* How to gracefully handle missing (`NaN`) values during string checks.

### Simple Explanation & Real-World Analogy
Imagine you are a librarian with a massive stack of book cards.
* A manual Python loop is like picking up each card one-by-one, reading the title, and checking if it starts with "The". It takes a long time.
* The `.str` accessor is like casting a magic spell over the entire catalog. Instantly, every card that starts with "The" glows blue. It is incredibly fast because it is vectorized (processed in parallel).

### Code Examples

First, let's create a DataFrame containing some messy student email and username records.


````python
import pandas as pd

data = {
    'Student_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice Smith', 'Bob Jones', 'Charlie Brown', 'David Miller', None],
    'Email': ['alice.smith@gmail.com', 'bjones@outlook.com', 'cbrown@gmail.com', 'dmiller@edu.net', 'missing_email']
}

df = pd.DataFrame(data)
print(df)
````

#### Checking Starts and Ends Patterns
Let's filter our students to find whose names start with "Alice" or whose emails end with ".com".

````python
# Check if Name starts with "Alice"
# We must use .str before calling startswith()
starts_alice = df['Name'].str.startswith('Alice')
print("Starts with Alice Mask:")
print(starts_alice)
````

Note how index 4 returned `None` (or `NaN`). Since David's name is `None`, Pandas does not know if it starts with "Alice" or not.

#### Pattern Matching with `.str.contains()` & Handling NaNs
If we want to find all Gmail users, we can search for the substring `"gmail"` in the `Email` column.

By default, if there are null values in the column, `.str.contains()` might return `NaN`, which causes errors when filtering. We use **`na=False`** (or `na=True`) to tell Pandas how to treat missing values.


````python
# Find all Gmail users
# na=False treats any NaN/None value as False so the filter doesn't crash
gmail_users = df[df['Email'].str.contains('gmail', na=False)]
print(gmail_users)
````

### Common Mistakes Beginners Make
1. **Forgetting the `.str` accessor:** Trying to call `df['Email'].contains('gmail')` directly. This will raise an `AttributeError` because the Series object itself does not have string methods; you must access the `.str` interface first: `df['Email'].str.contains('gmail')`.
2. **Crashing on Nulls:** Running `.str.contains()` on a column with missing values without setting the `na` parameter. If you use it as a mask, Pandas will raise a `ValueError` because boolean masks cannot contain null values. Always use `na=False` when filtering messy columns.

#### Exercise 1 (Easy)
Given the following Series of file names:
```python
files = pd.Series(['backup_2026.csv', 'report.pdf', 'data_new.csv', 'image.png', None])
```
Write Pandas code to filter and return only the files that end with the extension `".csv"`. Make sure your code does not crash on the `None` value.

````python
import pandas as pd
files = pd.Series(['backup_2026.csv', 'report.pdf', 'data_new.csv', 'image.png', None])

# Use .str.endswith() with na=False to handle the None value
csv_files = files[files.str.endswith('.csv', na=False)]
print(csv_files)
````
