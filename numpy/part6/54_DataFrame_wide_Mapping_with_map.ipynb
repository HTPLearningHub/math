# DataFrame-wide Mapping with `.map()`

Sometimes, you need to apply a transformation not just to a single column, but to **every single cell** of your entire DataFrame. For example, you might want to format all float columns to 2 decimal places, capitalize all text values, or prepend a character to every element.

In **Pandas 2.1+**, the method used for element-wise DataFrame-wide transformations is **`.map()`** (which replaced the older, now-deprecated `.applymap()` method).

Using `.map()` on a DataFrame applies your function to each individual element across the entire grid.

### Plain English Explanation & Real-World Analogy
Imagine you are an **interior decorator** walking into an empty office building.
*   **Series `.map()`** is like painting only one specific wall in the lobby.
*   **DataFrame `.map()`** is like going into *every single room* of the entire building and installing the exact same style of lightbulb. You are doing an element-by-element upgrade to the entire structure.

### Code Examples

Let's create a DataFrame with float numbers representing sales measurements across different quarters:


````python
import pandas as pd

sales = pd.DataFrame({
    'Q1_Revenue': [1024.556, 2048.112, 512.990],
    'Q2_Revenue': [1200.443, 2300.999, 450.123]
})
print("Original Sales:")
print(sales)
````

#### Formatting All Cells (DataFrame-wide `.map()`)
Let's round all numeric values in our DataFrame to exactly 2 decimal places and format them as currency strings:

````python
# Apply lambda element-wise to every cell in the DataFrame
formatted_sales = sales.map(lambda val: f"${val:.2f}")
print(formatted_sales)
````

#### Checking Element-wise Types
You can also use `.map()` to apply Python string manipulations across a text DataFrame. Let's convert all strings to lowercase:

````python
words_df = pd.DataFrame({
    'Group_A': ['APPLE', 'BANANA', 'CHERRY'],
    'Group_B': ['DOG', 'EAGLE', 'FROG']
})

# Convert all text cells to lowercase
lowercase_df = words_df.map(lambda x: x.lower())
print(lowercase_df)
````

### Common Mistakes Beginners Make
1.  **Using `.applymap()` in newer Pandas code**: While `.applymap()` still works as an alias in some environments, it has been officially deprecated. In modern Pandas (2.1+), always use **`.map()`** on both Series and DataFrames to keep your code clean and future-proof.
2.  **Using `.map()` on mixed-type DataFrames**: If you run `df.map(lambda x: x.lower())` on a DataFrame that contains both text and integer numbers, Python will crash with an `AttributeError` because integers do not have a `.lower()` method. Ensure your DataFrame has consistent data types before applying element-wise map operations!

#### Exercise 1 (Medium)
Given a DataFrame of mixed measurements:
```python
measurements = pd.DataFrame({
    'Width': [10.5, 20.0, 15.2],
    'Height': [5.1, 8.4, 12.0]
})
```
Use `.map()` to scale all measurements by multiplying every single cell by `2.5`.


````python
import pandas as pd

measurements = pd.DataFrame({
    'Width': [10.5, 20.0, 15.2],
    'Height': [5.1, 8.4, 12.0]
})

# Scale all cells by 2.5
scaled_df = measurements.map(lambda x: x * 2.5)
print(scaled_df)
````
