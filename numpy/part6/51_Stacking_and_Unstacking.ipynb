# Stacking and Unstacking (`.stack()` and `.unstack()`)

**Stacking and Unstacking** are reshaping operations specifically designed to work with DataFrames that have **MultiIndex** (hierarchical) rows or columns.

*   **`.stack()`**: "Tucks" or pivots the innermost column labels into the innermost row index. This makes the DataFrame **taller and narrower** (stacked).
*   **`.unstack()`**: The exact opposite. It "untucks" or pivots the innermost row index labels back into column headers. This makes the DataFrame **shorter and wider** (unstacked).

Understanding stack and unstack is vital when working with aggregated data (such as output from `.groupby()`), multi-dimensional datasets, or multi-tiered headers.

### Plain English Analogy
*   **Stacking**: Imagine you have several folders laid side-by-side on your desk (Columns). Stacking them means taking those folders and piling them on top of each other into a single vertical tower (Rows).
*   **Unstacking**: Taking a vertical stack of folders (Rows) and laying them out side-by-side across the desk (Columns).

### Code Examples

Let's create a DataFrame with multi-level columns representing Sales and Profits for different products:




````python
import pandas as pd

# Creating MultiIndex Columns
columns = pd.MultiIndex.from_tuples([
    ('Laptops', 'Sales'),
    ('Laptops', 'Profit'),
    ('Phones', 'Sales'),
    ('Phones', 'Profit')
])

df_multi = pd.DataFrame([
    [100, 20, 80, 15],
    [120, 25, 90, 18]
], index=['Store_A', 'Store_B'], columns=columns)

print("--- DataFrame with MultiIndex Columns ---")
print(df_multi)
````

#### Compression with `.stack()`
Let's collapse the innermost column level (the Sales/Profit headers) into row indices:

````python
# Stacking the innermost column level into rows
stacked = df_multi.stack()
print("--- Stacked DataFrame ---")
print(stacked)
````

*(Now, the row index has two levels: [Store, Metric], while the column headers have been simplified to just Laptops and Phones.)*

#### Expansion with `.unstack()`
Let's take our stacked DataFrame and unstack it back to columns:

````python
# Unstacking the innermost row index level back into columns
unstacked = stacked.unstack()
print("--- Unstacked Back to Wide ---")
print(unstacked)
````


### Common Pitfalls & Mistakes
*   **Missing values (NaNs) during unstacking**: If your row index levels do not contain perfect combinations for every single index, unstacking will create empty cells. Pandas automatically fills these with `NaN`. You can use the `fill_value` parameter inside `.unstack(fill_value=0)` to avoid NaNs.
*   **Losing track of levels**: MultiIndex arrays are zero-indexed from outer to inner. For rows: Level 0 is the outermost row, and Level 1 is the innermost. You can pass a level number to stack or unstack (e.g., `.unstack(level=0)`) to specify exactly which level you want to rotate.
