# Introduction to Pandas & The Series Object

In standard Python, lists are simple. In TypeScript, arrays store sequences of values. But in data science, we need to associate our data with meaningful labels (like dates, names, or IDs).

**Pandas** provides two main data structures:
1. **Series**: A 1-dimensional labeled array (like a single column in a spreadsheet).
2. **DataFrame**: A 2-dimensional labeled grid/table (like an entire Excel sheet).

In this module, we focus on the **Series**.
* **Why it matters:** Standard arrays only let you access elements by position indices (`0, 1, 2...`).
* A Pandas Series lets you use custom labels (like `"Monday"`, `"Tuesday"`, or country names) as indices. This makes searching and filtering extremely simple and human-readable.

### Simple Explanation with an Everyday Analogy
Imagine an **apartment building hallway**:
* **A standard array** is like a hallway where doors are only numbered: `Door 0`, `Door 1`, `Door 2`. To find your friend, you must count the doors.
* **A Pandas Series** is like a hallway where every door has a nameplate (index label). One door says `"Alex"`, another says `"John"`, and another says `"Sarah"`. You can go directly to `"Sarah"` without counting.

### 3. Code Examples
Let's see how to create and use a Series. First, we import `pandas` using the standard alias `pd`.




````python
import pandas as pd

# Creating a basic Series from a list
temperatures = pd.Series([22.5, 25.0, 19.8, 28.1])
print("Basic Series:", temperatures)
````

*Note:* The left side shows the default integer index (`0, 1, 2, 3`). The bottom shows the metadata (data type is `float64`).


#### Custom Index Labels
We can pass a list of custom labels to the `index` parameter:


````python
import pandas as pd

# Series with a custom string index
sales = pd.Series([100, 150, 85], index=["Monday", "Tuesday", "Wednesday"])
print("Series with custom index:", sales)
````

#### Accessing and Updating Values
We can access values using `.loc` (location by label) or `.iloc` (integer location by position).

````python
import pandas as pd

sales = pd.Series([100, 150, 85], index=["Monday", "Tuesday", "Wednesday"])

# Accessing by label using .loc
print("Tuesday Sales:", sales.loc["Tuesday"])

# Accessing by position using .iloc
print("First day sales:", sales.iloc[0])

# Updating a value
sales.loc["Monday"] = 120
print("Updated Series:", sales)
````

### 4. Common Mistakes Beginners Make
1. **Mismatch of sizes**: When creating a Series with a custom index, your list of labels **must be the same length** as your list of values. If they do not match, Pandas will throw a `ValueError`.
2. **Confusing `.loc` and `.iloc`**:
   * Use `.loc` when you want to search by the **name** of the label.
   * Use `.iloc` when you want to search by the **integer position** (`0, 1, 2...`).

#### Exercise 2 (Medium)
You are given this Series of monthly cloud costs:

1. Print the cost for `"July"` using `.loc`.

#### Exercise 3 (Hard)
Create a Series representing daily steps walked: `[8000, 12000, 4500, 15000, 9800]` with index labels representing days: `["Mon", "Tue", "Wed", "Thu", "Fri"]`. Filter this Series to only show days where steps were greater than `10000` [352].


````python
import pandas as pd
steps = pd.Series([8000, 12000, 4500, 15000, 9800], index=["Mon", "Tue", "Wed", "Thu", "Fri"])

# Filtering using boolean indexing
active_days = steps[steps > 10000]
print(active_days)
````

*Explanation:* We pass the condition `steps > 10000` inside square brackets. Pandas evaluates this and keeps only the rows that are `True`.
