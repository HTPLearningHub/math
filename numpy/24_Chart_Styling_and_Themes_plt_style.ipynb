# Chart Styling and Themes (`plt.style`)

By default, Matplotlib plots look very simple with a white background and basic colors. In professional data science, presentation is everything.
In Angular, you use CSS or pre-made UI themes (like Angular Material) to make your app look modern. In Matplotlib, we use **Stylesheets (Themes)**.

With just one line of code, you can change the background, gridlines, fonts, and colors of your entire plot to match professional standards (like the style of *FiveThirtyEight* website or *ggplot* from R).

### Simple Explanation with an Everyday Analogy
Imagine you are building a website:
*   **Without Stylesheets**: You have raw, black-and-white HTML with no CSS. It works, but it looks outdated.
*   **With Stylesheets (`plt.style`)**: It is like adding a CSS framework (like Tailwind or Bootstrap). All colors, margins, and borders update instantly.

### Code Examples

Let's see what styles are available and how to apply them.

#### Checking Available Styles
You can see all built-in styles in Matplotlib using `plt.style.available`.

```python
import matplotlib.pyplot as plt

# Show a few available styles
print("Available styles (first 5):", plt.style.available[:5])
```

#### Applying a Style globally
To apply a theme, use `plt.style.use('style_name')`. This changes the theme for **all** plots created after this line.

```python
import numpy as np
import matplotlib.pyplot as plt

# Apply ggplot style (popular style from R programming)
plt.style.use('ggplot')

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("ggplot Style Theme")
# In our scripts we save or show, but the style is globally active now!
```

#### Temporary Styling using Context Manager
If you only want a specific style for *one* plot without changing the global settings, use the `with` statement:

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.cos(x)

# Use dark background ONLY for this block
with plt.style.context('dark_background'):
    plt.plot(x, y, color='cyan')
    plt.title("Temporary Dark Theme")
```

### Common Mistakes Beginners Make
1.  **Forgetting that `plt.style.use()` is global**: If you call `plt.style.use('dark_background')` at the top of your Jupyter Notebook, all your next plots will be black, even if you do not want them to be. Use `plt.style.use('default')` to reset it.
2.  **Using a style that is not installed**: If you make a typo like `plt.style.use('gg-plot')` instead of `'ggplot'`, Python will throw an error.

#### Exercise 2 (Medium)
Create a line plot of $y = x^2$ for $x$ from -5 to 5. Apply the `'fivethirtyeight'` theme temporarily using a context manager, and make the line color red.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 50)
y = x ** 2

with plt.style.context('fivethirtyeight'):
    plt.plot(x, y, color='red')
    plt.title("FiveThirtyEight Theme")
```
