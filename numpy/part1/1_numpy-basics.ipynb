# Numpy Basics

in Data Science, Artificial Intelligence (AI), and Machine Learning, you must process **millions of numbers**.
Standard Python lists are not built for this kind of heavy workload. They are too slow.
NumPy is the foundational library (a collection of pre-written code that you can reuse) for all data science and AI
work in Python. Learning NumPy is the absolute first step if you want to work with data or AI.

## The Python List: A Scattered Shopping List

Imagine you have a shopping list. The items on your list are scattered all over a giant supermarket.

* To get item 1 (apples), you walk to row 5.
* To get item 2 (shampoo), you walk to row 12.
* To get item 3 (a book), you walk to row 2.

This is how a **Python List** works in your computer's memory. It uses **non-contiguous memory**. This means the numbers are scattered in
different memory locations. Your computer has to use "pointers" (memory addresses) to jump around from one number to another. This jumping
around takes time.

Also, Python lists are dynamically typed. This means a single list can hold an integer, a string, and a boolean all at once. Because of this,
every time Python reads an item, it must ask: *"What type of data is this?"* This constant check makes it slow.

## The NumPy Array: The Factory Conveyor Belt

Now, imagine a high-speed factory conveyor belt. On this belt, there are identical, blue boxes placed right next to each other.

* Every box is exactly the same size.
* Every box contains the exact same type of item (for example, only 32-bit integers).
* There is no gaps between the boxes.

This is a **NumPy Array** (specifically, an `nd array` or N-Dimensional Array).

Because the boxes are identical and adjacent (right next to each other), your computer can read them in one super-fast sweep. This is called
**contiguous memory**. The computer does not have to jump around or do any type checking. It already knows every item is the same type.

Now, we import NumPy in our Python file. We give it a short nickname `np` so we do not have to type `numpy` every time. This nickname is a universal standard.

```python
# Import the NumPy library
import numpy as np

# Print the installed version to check if it works
print("The numpy version is " + np.__version__)
```

#### Example 1.1: Python List vs. NumPy Array (Multiplication)
Let's see what happens when we try to multiply a sequence of numbers by 2.

```python
# 1. Using a standard Python list
python_list = [1, 2, 3, 4]
list_result = python_list * 2
print("Python List Result:", list_result)

# 2. Using a NumPy array
numpy_array = np.array([1, 2, 3, 4])
array_result = numpy_array * 2
print("NumPy Array Result :", array_result)
```


**Why this works:**
* When you multiply a Python list by 2, Python thinks you want to duplicate (repeat) the list.
* When you multiply a NumPy array by 2, NumPy performs a **vectorized operation**. This means it applies the multiplication to
    **every single element** in the array automatically, without a loop. This is done using your computer's CPU vector instructions
    (called **SIMD**: Single Instruction, Multiple Data).


#### Example 1.2: Python List vs. NumPy Array (Adding a Number)
In Angular, if you wanted to add 10 to every number in an array, you would use `.map(x => x + 10)`. Let's see how Python lists and NumPy do this.


```python
# 1. Using a Python list (We must use a loop or list comprehension)
python_list = [1, 2, 3, 4, 5]
list_result = [x + 10 for x in python_list]
print("Python List + 10:", list_result)

# 2. Using a NumPy array (Direct scalar addition)
numpy_array = np.array([1, 2, 3, 4, 5])
array_result = numpy_array + 10
print("NumPy Array + 10 :", array_result)
```

**Why this works:**
* Adding `10` to `numpy_array` is a **scalar operation**. A "scalar" is just a single number. NumPy virtually expands that single number `10`
    and adds it to every element in the array instantly. No slow Python loop is needed!

#### Mistake 1: Expecting math operations to work on standard Python lists

```python
# WRONG: This will throw an error or do duplication, not math!
my_list = [1, 2, 3]
print(my_list + 5)  # TypeError: can only concatenate list (not "int") to list
```

**Correction:** Convert the list to a NumPy array first.

```python
my_array = np.array([1, 2, 3])
print(my_array + 5)  # Output: [6 7 8]
```

#### Mistake 2: Forgetting that print(array) does not show commas
When you print a Python list, you see commas: `[1, 2, 3]`.
When you print a NumPy array, you do not see commas, only spaces: `[1 2 3]`.
Do not be confused! It is still a sequence of numbers, just printed differently by NumPy.
