# Creating NumPy Arrays

When you analyze data or build machine learning models, you rarely enter millions of numbers by hand.

* You might need a range of coordinates (like time points `0, 0.1, 0.2...`).
* You might need an empty array to store predictions or results before a computation starts.
* You might need to copy an array to modify it without breaking your original data.

Knowing how to initialize (create and prepare) these arrays correctly is essential for memory efficiency and avoiding buggy code.

## Simple Explanation with an Everyday Analogy

#### np.zeros() and np.ones(): Egg Cartons

Imagine you are baking. You buy an empty egg carton. It has 12 empty cups. Each cup is a placeholder.
* `np.zeros(12)` is like an egg carton where every cup is filled with a dummy egg of value `0`.
* `np.ones(12)` is like a carton where every cup is filled with a dummy egg of value `1`.

#### np.arange(): The Staircase

Imagine a custom-built staircase.
* You start at step 0 (Start).
* You stop *before* step 10 (Stop - exclusive, which means we do not include it).
* Each step goes up by 2 (Step size).

This is `np.arange(0, 10, 2)`. It creates steps of: `0, 2, 4, 6, 8`.


#### np.linspace(): Dividing a Ruler

Imagine you have a piece of wood that is 1 meter long. You want to cut it into **5 equal parts**.
You do not care about the "step size" in centimeters; you just want exactly 5 evenly-spaced marks from 0 to 1 meter.
This is `np.linspace(0, 1, 5)`. It will calculate the marks for you: `0.0, 0.25, 0.5, 0.75, 1.0`.

#### Copy vs. View: The Photocopy vs. The Web Shortcut
If you have a document on your desk:
* **Shallow Copy (View / Reference)**: This is like creating a shortcut icon on your desktop. If you click the shortcut
    and edit the file, you are editing the original file! In Python, writing `b = a` creates a reference. Changing `b`
    will change `a`.
* **Deep Copy (np.copy())**: This is like making a physical photocopy of a document. If you draw on the photocopy
    with a red pen, the original document on your desk remains clean.




#### Example: Basic Sequences with np.arange() and np.linspace()

```python
import numpy as np

# Create a range of numbers: start at 1, stop before 10, step by 2
range_arr = np.arange(1, 10, 2)
print("np.arange(1, 10, 2):", range_arr)

# Create 5 evenly spaced numbers between 0 and 1 (inclusive)
linspace_arr = np.linspace(0, 1, 5)
print("np.linspace(0, 1, 5):", linspace_arr)
```


**Why this works:**
* `np.arange(start, stop, step)` calculates numbers step-by-step. The `stop` value is **exclusive** (never included).
* `np.linspace(start, stop, num)` calculates exactly `num` elements. The `stop` value is **inclusive** by default.
    It divides the range into equal segments.


#### Example: Placeholders (zeros, ones, full, empty)

```python
import numpy as np

# Array of 4 zeros
zeros_arr = np.zeros(4)
print("np.zeros(4):", zeros_arr)

# Array of 4 ones
ones_arr = np.ones(4)
print("np.ones(4) :", ones_arr)

# Array of 4 sevens
full_arr = np.full(4, 7)
print("np.full(4, 7):", full_arr)

# Empty array of 3 elements (warning: values are "garbage")
empty_arr = np.empty(3)
print("np.empty(3)  :", empty_arr)
```

**Why this works:**

* `np.zeros()` and `np.ones()` create arrays prefilled with floats `0.0` or `1.0`.
* `np.full()` fills the array with your custom value.
* `np.empty()` does **not** initialize or clean the memory. It just grabs whatever random data was left in that
    memory location by other programs. This is extremely fast because the computer doesn't waste time writing zeros.
    But you must fill it with your own values before reading it!


#### The "b = a" Trap vs. np.copy()
In TypeScript, if you write `let b = a` for an array, both point to the same array in memory.
Python and NumPy work the same way!


```python
import numpy as np

# Original array
a = np.array([1, 2, 3])

# Reference assignment (Trap!)
b = a

# Proper copy
c = a.copy()

# Modify b
b[0] = 99

print("Original array a:", a)
print("Reference array b:", b)
print("True copy array c:", c)
```

**Why this works:**

* `b = a` does **not** create a new array. It just creates a new name `b` pointing to the exact
    same memory block as `a`. Modifying `b` changes the values in `a`.
* `a.copy()` allocates a completely new block of memory and copies the data into it.
    Modifying `c` would have no effect on `a`.


#### Mistake 1: Swapping np.arange() and np.linspace() arguments

```python
# WRONG: You want numbers from 0 to 10 with step 2, but write linspace
arr = np.linspace(0, 10, 2)  # This gives only 2 elements: [0. 10.]
print(arr)
```

**Correction:**
* Use `np.arange(0, 11, 2)` if you want to specify the step size (2).
* Use `np.linspace(0, 10, 6)` if you want to specify the total number of elements (6 elements will give `[0. 2. 4. 6. 8. 10.]`).


#### Mistake 2: Using np.empty() and expecting it to be empty (all zeros)

Some beginners use `np.empty()` thinking it creates an array of zeros. As shown above, it contains random, unpredictable "garbage" values. If you need zeros, always use `np.zeros()`.

#### Exercise 2
1. Use `np.arange` to create an array of all odd numbers between `1` and `15` (inclusive).
2. Use `np.linspace` to create an array of `11` numbers evenly spaced between `0.0` and `1.0`. Print both arrays.

```python
import numpy as np

# 1. Odd numbers between 1 and 15 (inclusive)
# We stop at 16 because the stop value is exclusive.
odds = np.arange(1, 16, 2)
print("Odd numbers:", odds)

# 2. 11 numbers between 0.0 and 1.0
lin_spaced = np.linspace(0.0, 1.0, 11)
print("Linspace numbers:", lin_spaced)
```

**Explanation:**
* For `arange`, we start at `1` and step by `2`. To include `15`, we must set the stop bound higher than `15` (like `16` or `17`).
* For `linspace`, we want exactly `11` numbers between `0.0` and `1.0` [52, 105]. NumPy calculates the exact step (`0.1`) automatically.


#### Exercise

Write a Python script where you:
1. Create a NumPy array `original` containing `[10, 20, 30, 40]`.
2. Create a reference named `ref_view` using `ref_view = original`.
3. Create a deep copy named `deep_copy` using `deep_copy = original.copy()`.
4. Change the last element of `ref_view` to `999`.
5. Change the first element of `deep_copy` to `111`.
6. Print all three arrays. Write a comment explaining why each array looks the way it does.

```python
import numpy as np

original = np.array([10, 20, 30, 40])
ref_view = original
deep_copy = original.copy()

# Modify values
ref_view[-1] = 999  # -1 means the last element in Python!
deep_copy[0] = 111

print("Original :", original)
print("ref_view :", ref_view)
print("deep_copy:", deep_copy)
```

**Explanation:**
* `ref_view` is just an alias (pointer) for `original`. Changing `ref_view[-1]` directly changes `original[-1]`.
* `deep_copy` is a completely independent copy stored in a different memory location. Changing `deep_copy[0]`
    does not affect `original`.
