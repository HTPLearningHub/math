# Modern Random Number Generation (Simulations and Choices)

In simulations, video games, data science, and testing, we constantly need random numbers.
In JavaScript/TypeScript, you only have `Math.random()`, which returns a decimal number between 0 and 1. If you want a random integer, you have to write math formulas yourself:
```typescript
const randomInt = Math.floor(Math.random() * (max - min) + min);
```
NumPy has a complete, highly optimized sub-library for random calculations: `np.random`.

*Crucial Modern NumPy Practice:*
If you search Google, you will see old tutorials using functions like `np.random.randint()` or `np.random.seed()`. This is **legacy** (old) code.
In modern NumPy (since version 1.17), the correct way is to create a **Generator object** using `np.random.default_rng()`. 
This new approach is faster, thread-safe, and uses better algorithms. We will use this modern standard exclusively.

We will learn how to:
1.  Generate random integers and float numbers in custom ranges.
2.  Use a **Seed** to make "random" numbers repeatable (essential for debugging!).
3.  **Shuffle** an array in-place.
4.  Make **Random Choices** from an array of options (with or without replacement).


### Simple Explanation with an Everyday Analogy
Think of the random number generator (`rng`) as a **casino dealer with a deck of cards**:
*   **Dealer (`default_rng`)**: This is the dealer you hire.
*   **Seed**: Imagine the dealer shuffles cards using a book of instructions. If you give the dealer the number code `42` (the Seed), the dealer will shuffle and deal the cards in the **exact same order** every time you play. This is amazing for science because another scientist can use the same seed and get the exact same results.
*   **Shuffle**: The dealer scrambles the cards already in your hand. No cards are lost, but their order changes.
*   **Choice**: The dealer lays out cards (e.g. apple, orange, banana) and lets you draw cards. If `replace=True`, you draw a card, write it down, and put it back in the deck before drawing again (you can get the same card twice). If `replace=False`, you keep the card (you cannot draw it twice).


#### Initializing the Generator and Generating Numbers
Let's see how to generate integers and floats.

````python
import numpy as np

# 1. Create the Generator (rng)
# We set seed=42 so that the outputs are always the same when we run the code
rng = np.random.default_rng(seed=42)

# 2. Generate 5 random integers between 1 and 9 (10 is exclusive!)
random_ints = rng.integers(low=1, high=10, size=5)
print("5 Random Integers [1, 9]:", random_ints)

# 3. Generate a 2D grid of random integers (3 rows, 2 columns)
grid = rng.integers(low=10, high=100, size=(3, 2))
print("Random 3x2 Grid:\n", grid)

# 4. Generate 3 random floats between 0.0 and 1.0 (uniform distribution)
random_floats = rng.uniform(low=0.0, high=1.0, size=3)
print("3 Random Floats:", random_floats)
````

#### Shuffling Arrays
Shuffling reorganizes the elements of an array. 
*Note*: The `.shuffle()` method modifies your array **in-place** (it changes the original array directly) and returns `None`.

````python
import numpy as np

rng = np.random.default_rng(seed=42)

arr = np.array([1, 2, 3, 4, 5])
print("Array before shuffle:", arr)

# Shuffle in place
rng.shuffle(arr)
print("Array after shuffle:", arr)
````

#### Random Choices (Sampling)
You can pick random elements from an array of options (numbers or strings).
*   `replace=True` (default): Elements can be picked multiple times.
*   `replace=False`: Elements are picked without replacement (each element can be chosen at most once).


````python
import numpy as np

rng = np.random.default_rng(seed=42)

fruits = np.array(["Apple", "Orange", "Banana", "Kiwi"])

# Pick 3 random fruits (can repeat)
draws_repeat = rng.choice(fruits, size=3, replace=True)
print("Choices with repeat allowed:", draws_repeat)

# Pick 2 random fruits (cannot repeat)
draws_no_repeat = rng.choice(fruits, size=2, replace=False)
print("Choices with NO repeat allowed:", draws_no_repeat)
````

### Common Mistakes Beginners Make
1.  **Thinking `.shuffle()` returns a new array**: Beginners write `shuffled_arr = rng.shuffle(arr)`. This makes `shuffled_arr` equal to `None`, because the shuffle happens directly inside `arr`. Always call `rng.shuffle(arr)` first, and then use `arr`.
2.  **Using `replace=False` with a size larger than the array**: If you have 3 fruits, and you ask NumPy to pick 5 fruits with `replace=False` (no repeats), NumPy will crash with a `ValueError` because it runs out of items to pick!
3.  **Using legacy functions**: Writing `np.random.randint()` instead of the modern `rng.integers()`. Always initialize your generator with `np.random.default_rng()`.


#### Exercise 1
1. Initialize a modern random generator with a seed of `10`.
2. Generate a 1D array of `4` random integers between `1` and `100` (inclusive of 100!).
3. Print the array.

````python
import numpy as np

# 1. Init generator with seed 10
rng = np.random.default_rng(seed=10)

# 2. Generate 4 integers between 1 and 100.
# Because the upper limit is exclusive, we must set high to 101!
random_nums = rng.integers(low=1, high=101, size=4)
print(random_nums)
````

*Explanation*: `high=101` makes `100` the highest possible integer. Using seed `10` guarantees these exact 4 numbers.



#### Exercise 2
You are building a game. You have a list of player names:
```python
players = np.array(["Alice", "Bob", "Charlie", "David", "Eve"])
```
1. Initialize a generator with seed `7`.
2. Shuffle the players to randomize their playing order.
3. Randomly select **one** MVP player from the list using `.choice()`.


````python
import numpy as np

rng = np.random.default_rng(seed=7)
players = np.array(["Alice", "Bob", "Charlie", "David", "Eve"])

# Shuffle players in-place
rng.shuffle(players)
print("Randomized order:", players)

# Pick one MVP
mvp = rng.choice(players)
print("MVP:", mvp)
````

*Explanation*: `rng.shuffle()` shuffled the array directly. `rng.choice(players)` picked one random name from the shuffled list.


#### Exercise 3
You want to simulate a coin toss.
1. Initialize a generator with seed `123`.
2. Generate an array of `1000` integers where `0` represents "Tails" and `1` represents "Heads".
3. Use an aggregate function from Module 7 to count how many times "Heads" occurred.
4. Calculate the percentage of Heads (e.g. if Heads occurred 512 times, the percentage is 51.2%).


````python
import numpy as np

rng = np.random.default_rng(seed=123)

# 1000 coin flips: 0 or 1 (so high must be 2!)
flips = rng.integers(low=0, high=2, size=1000)

# Since Heads is 1 and Tails is 0, the sum of the array is the count of Heads!
heads_count = np.sum(flips)
percentage = (heads_count / 1000) * 100

print("Heads occurred:", heads_count, "times")
print("Percentage of Heads:", percentage, "%")
````

*Explanation*: Summing the array of 0s and 1s beautifully counts the 1s (Heads). In this random run, Heads appeared 489 times (48.9%), which is very close to the theoretical 50% probability!
